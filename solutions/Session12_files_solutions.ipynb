### تمرین ۲

فایل `products.csv` با ستون‌های `name`، `price` و `qty` برای ۴ کالا بسازید. سپس آن را بخوانید و ارزش کل موجودی (`price * qty`) را حساب کنید.

</div>

```python
import csv

products = [
    {"name": "pen", "price": 45000, "qty": 120},
    {"name": "book", "price": 280000, "qty": 15},
    {"name": "bag", "price": 950000, "qty": 4},
    {"name": "ruler", "price": 30000, "qty": 60},
]
with open("products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "price", "qty"])
    writer.writeheader()
    writer.writerows(products)

total = 0
with open("products.csv", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        total += int(row["price"]) * int(row["qty"])
print(f"inventory value: {total:,}")
```

### تمرین ۳

دیکشنری تنظیمات یک برنامه (مثلاً زبان، اندازهٔ فونت و حالت تیره) را در `settings.json` ذخیره کنید، سپس آن را بخوانید، اندازهٔ فونت را ۲ واحد بزرگ کنید و دوباره ذخیره کنید.

</div>

```python
import json

settings = {"language": "fa", "font_size": 14, "dark_mode": True}
with open("settings.json", "w", encoding="utf-8") as f:
    json.dump(settings, f, ensure_ascii=False, indent=2)

with open("settings.json", encoding="utf-8") as f:
    settings = json.load(f)
settings["font_size"] += 2

with open("settings.json", "w", encoding="utf-8") as f:
    json.dump(settings, f, ensure_ascii=False, indent=2)
print(settings)
```

### تمرین ۴

محتوای فایل `students.csv` (ساخته‌شده در درس) را بخوانید و آن را به شکل لیستی از دیکشنری‌ها در `students.json` ذخیره کنید (نمره را به عدد تبدیل کنید).

</div>

```python
import csv
import json

with open("students.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
for r in rows:
    r["score"] = float(r["score"])

with open("students.json", "w", encoding="utf-8") as f:
    json.dump(rows, f, ensure_ascii=False, indent=2)
print(rows)
```

### تمرین ۵

تابع `search_file(path, word)` بنویسید که شمارهٔ خطوطی از فایل را که شامل کلمهٔ موردنظر هستند برگرداند.

</div>

```python
def search_file(path, word):
    found = []
    with open(path, encoding="utf-8") as f:
        for number, line in enumerate(f, start=1):
            if word.lower() in line.lower():
                found.append(number)
    return found

with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("Python is fun\nI love data\npython again\n")
print(search_file("notes.txt", "python"))
```
