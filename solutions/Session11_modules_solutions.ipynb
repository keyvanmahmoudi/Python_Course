### تمرین ۱

با ماژول `random` و `seed(7)` یک رمز تصادفی ۸ کاراکتری از حروف و ارقام بسازید (راهنما: `string.ascii_letters` و `string.digits`).

</div>

```python
import random
import string

random.seed(7)
chars = string.ascii_letters + string.digits
password = "".join(random.choice(chars) for _ in range(8))
print(password)
```

### تمرین ۲

تاریخ تولد میلادی خود را با `date` بسازید و حساب کنید تا امروز چند روز زنده بوده‌اید و تولد بعدی‌تان چند روز دیگر است.

</div>

```python
from datetime import date

birthday = date(2000, 3, 21)
today = date.today()
print("days lived:", (today - birthday).days)

next_bd = birthday.replace(year=today.year)
if next_bd < today:
    next_bd = next_bd.replace(year=today.year + 1)
print("days to next birthday:", (next_bd - today).days)
```

### تمرین ۳

با `Counter` سه حرف پرتکرار جملهٔ `"data science and machine learning"` را (بدون احتساب فاصله) پیدا کنید.

</div>

```python
from collections import Counter

sentence = "data science and machine learning"
counts = Counter(sentence.replace(" ", ""))
print(counts.most_common(3))
```

### تمرین ۴

یک ماژول به نام `converters.py` با دو تابع `km_to_mile` و `kg_to_pound` بسازید و آن را import و امتحان کنید.

</div>

```python
from pathlib import Path

Path("converters.py").write_text('''
def km_to_mile(km):
    return km * 0.621371


def kg_to_pound(kg):
    return kg * 2.20462
''', encoding="utf-8")

import converters
print(converters.km_to_mile(10))
print(converters.kg_to_pound(70))
```

### تمرین ۵

۱۰۰۰ بار یک تاس را با `random` (و `seed(1)`) پرتاب کنید و با `Counter` فراوانی هر عدد را چاپ کنید. آیا تقریباً برابرند؟

</div>

```python
import random
from collections import Counter

random.seed(1)
rolls = [random.randint(1, 6) for _ in range(1000)]
print(sorted(Counter(rolls).items()))
```
