### تمرین ۱

### تمرین ۲

یک عدد سه‌رقمی (مثلاً `572`) را در متغیر بگذارید و با `//` و `%` رقم صدگان، دهگان و یکان آن را جدا چاپ کنید.

</div>

```python
n = 572
hundreds = n // 100
tens = (n // 10) % 10
ones = n % 10
print(hundreds, tens, ones)
```

### تمرین ۳

برنامه‌ای بنویسید که تعداد ثانیه را بگیرد (مثلاً `3725`) و آن را به شکل ساعت، دقیقه و ثانیه چاپ کند.

</div>

```python
total_seconds = 3725
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(hours, "h", minutes, "m", seconds, "s")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

### تمرین ۴

دمای هوا به سانتی‌گراد را در یک متغیر بگذارید و آن را به فارنهایت تبدیل کنید ($F = C \times 9/5 + 32$).

</div>

```python
celsius = 25
fahrenheit = celsius * 9 / 5 + 32
print(celsius, "C =", fahrenheit, "F")
```

### تمرین ۵

مقدار دو متغیر `a = 10` و `b = 20` را بدون استفاده از متغیر سوم جابه‌جا کنید و نتیجه را چاپ کنید.

</div>

```python
a = 10
b = 20
a, b = b, a
print("a =", a, "b =", b)
```
