### تمرین ۱

کلاس `Animal` با متد `speak()` بنویسید و سه کلاس فرزند `Dog`، `Cat` و `Cow` بسازید که هر کدام صدای خود را برگردانند. با یک حلقه همهٔ حیوانات را به صدا درآورید.

</div>

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."


class Dog(Animal):
    def speak(self):
        return "Woof"


class Cat(Animal):
    def speak(self):
        return "Meow"


class Cow(Animal):
    def speak(self):
        return "Moo"


for animal in [Dog("Rex"), Cat("Tom"), Cow("Daisy")]:
    print(f"{animal.name}: {animal.speak()}")
```

### تمرین ۲

کلاس `SavingsAccount` را از `BankAccount` (جلسهٔ ۱۴) به ارث ببرید و متد `add_interest(rate)` اضافه کنید. همچنین `withdraw` را طوری بازنویسی کنید که حداقل موجودی ۱۰۰ هزار تومان حفظ شود.

</div>

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("insufficient funds")
        self.balance -= amount


class SavingsAccount(BankAccount):
    MIN_BALANCE = 100_000

    def add_interest(self, rate):
        self.balance += self.balance * rate

    def withdraw(self, amount):
        if self.balance - amount < self.MIN_BALANCE:
            raise ValueError("minimum balance must be kept")
        super().withdraw(amount)


acc = SavingsAccount("Sara", 1_000_000)
acc.add_interest(0.2)
acc.withdraw(500_000)
print(acc.balance)
try:
    acc.withdraw(700_000)
except ValueError as e:
    print("Error:", e)
```

### تمرین ۳

کلاس `Money` با ویژگی `amount` بنویسید که `__add__`، `__eq__`، `__lt__` و `__str__` (با جداکنندهٔ هزارگان و واحد تومان) داشته باشد. لیستی از چند `Money` را با `sorted` مرتب کنید.

</div>

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __eq__(self, other):
        return self.amount == other.amount

    def __lt__(self, other):
        return self.amount < other.amount

    def __str__(self):
        return f"{self.amount:,} Toman"


prices = [Money(250_000), Money(45_000), Money(1_200_000)]
print([str(m) for m in sorted(prices)])
print(prices[0] + prices[1])
print(Money(10) == Money(10))
```

### تمرین ۴

کلاس `Product` با ویژگی `price` بنویسید که با `@property` و setter اجازهٔ قیمت منفی ندهد، و یک property فقط‌خواندنی `price_with_tax` (۱۰٪ مالیات) داشته باشد.

</div>

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("price cannot be negative")
        self._price = value

    @property
    def price_with_tax(self):
        return self._price * 1.1


p = Product("book", 280_000)
print(p.price_with_tax)
try:
    p.price = -5
except ValueError as e:
    print("Error:", e)
```
