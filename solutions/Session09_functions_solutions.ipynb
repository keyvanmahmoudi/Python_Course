### تمرین ۱

تابع `is_even(n)` بنویسید که اگر `n` زوج بود `True` و وگرنه `False` برگرداند. آن را روی اعداد ۱ تا ۵ امتحان کنید.

</div>

```python
def is_even(n):
    return n % 2 == 0

for i in range(1, 6):
    print(i, is_even(i))
```

### تمرین ۲

تابع `celsius_to_fahrenheit(c)` و تابع عکس آن `fahrenheit_to_celsius(f)` را بنویسید و درستی‌شان را با تبدیل رفت و برگشت عدد ۳۷ بررسی کنید.

</div>

```python
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5 / 9

f = celsius_to_fahrenheit(37)
print(f, fahrenheit_to_celsius(f))
```

### تمرین ۳

تابع `count_vowels(text)` بنویسید که تعداد حروف صدادار انگلیسی (`aeiou`، بزرگ یا کوچک) را در متن برگرداند.

</div>

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

print(count_vowels("Machine Learning"))
```

### تمرین ۴

تابع `power(base, exp=2)` بنویسید که بدون استفاده از `**` و با حلقه، توان را حساب کند. با و بدون آرگومان دوم امتحانش کنید.

</div>

```python
def power(base, exp=2):
    result = 1
    for _ in range(exp):
        result *= base
    return result

print(power(5))
print(power(2, 10))
```

### تمرین ۵

تابع `stats(numbers)` بنویسید که میانگین، بیشینه و تعداد اعداد بالای میانگین را با هم (به شکل تاپل) برگرداند.

</div>

```python
def stats(numbers):
    avg = sum(numbers) / len(numbers)
    above = 0
    for n in numbers:
        if n > avg:
            above += 1
    return avg, max(numbers), above

avg, highest, above = stats([12, 18, 15, 20, 9])
print(avg, highest, above)
```
