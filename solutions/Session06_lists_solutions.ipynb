### تمرین ۱

لیست `[23, 7, 15, 42, 8, 16]` را بگیرید و بدون استفاده از `max` و `min`، بزرگ‌ترین و کوچک‌ترین عضو را با حلقه پیدا کنید.

</div>

```python
numbers = [23, 7, 15, 42, 8, 16]
largest = numbers[0]
smallest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n
    if n < smallest:
        smallest = n
print("max:", largest, "min:", smallest)
```

### تمرین ۲

از لیست `[3, 8, 3, 1, 8, 9, 1]` لیست جدیدی بسازید که اعضای تکراری را نداشته باشد و ترتیب اولین وقوع حفظ شود.

</div>

```python
numbers = [3, 8, 3, 1, 8, 9, 1]
unique = []
for n in numbers:
    if n not in unique:
        unique.append(n)
print(unique)
```

### تمرین ۳

دو لیست `names = ["Ali", "Sara", "Reza", "Mina"]` و `scores = [14, 19, 11, 17]` داریم. نام کسانی را که نمرهٔ بالای میانگین گرفته‌اند چاپ کنید.

</div>

```python
names = ["Ali", "Sara", "Reza", "Mina"]
scores = [14, 19, 11, 17]
average = sum(scores) / len(scores)
for name, score in zip(names, scores):
    if score > average:
        print(name, score)
```

### تمرین ۴

لیست `[1, 2, 3, 4, 5, 6]` را به دو لیست `evens` و `odds` تقسیم کنید.

</div>

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = []
odds = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)
print("evens:", evens)
print("odds:", odds)
```

### تمرین ۵

ماتریس `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]` را در نظر بگیرید. مجموع قطر اصلی (۱ + ۵ + ۹) و مجموع هر ستون را حساب کنید.

</div>

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
diagonal = 0
for i in range(len(matrix)):
    diagonal += matrix[i][i]
print("diagonal sum:", diagonal)

for col in range(3):
    col_sum = 0
    for row in range(3):
        col_sum += matrix[row][col]
    print(f"column {col} sum:", col_sum)
```
