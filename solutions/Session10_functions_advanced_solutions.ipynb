### تمرین ۱

تابع `longest(*words)` بنویسید که هر تعداد کلمه بگیرد و بلندترین را برگرداند.

</div>

```python
def longest(*words):
    return max(words, key=len)

print(longest("ai", "python", "data"))
```

### تمرین ۲

با List Comprehension لیست مکعب اعداد فرد بین ۱ تا ۱۵ را بسازید.

</div>

```python
cubes = [x ** 3 for x in range(1, 16) if x % 2 == 1]
print(cubes)
```

### تمرین ۳

لیست `products = [("pen", 45000), ("book", 280000), ("bag", 950000), ("ruler", 30000)]` را بر اساس قیمت نزولی مرتب کنید و با Dict Comprehension دیکشنری کالاهای زیر ۱۰۰ هزار تومان را بسازید.

</div>

```python
products = [("pen", 45000), ("book", 280000), ("bag", 950000), ("ruler", 30000)]
print(sorted(products, key=lambda p: p[1], reverse=True))
cheap = {name: price for name, price in products if price < 100000}
print(cheap)
```

### تمرین ۴

تابع بازگشتی `fib(n)` بنویسید که n-امین عدد فیبوناچی را برگرداند (`fib(0) = 0`، `fib(1) = 1`) و ۱۰ عدد اول را با comprehension چاپ کنید.

</div>

```python
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print([fib(i) for i in range(10)])
```

### تمرین ۵

تابع `make_report(title, **fields)` بنویسید که عنوان و هر تعداد فیلد نام‌دار بگیرد و یک رشتهٔ چندخطی مرتب (هر فیلد در یک خط) برگرداند.

</div>

```python
def make_report(title, **fields):
    lines = [title, "-" * len(title)]
    for key, value in fields.items():
        lines.append(f"{key}: {value}")
    return "\n".join(lines)

print(make_report("Student", name="Sara", score=19, city="Sari"))
```
