### تمرین ۱

مجموع اعداد زوج بین ۱ تا ۱۰۰ را با حلقهٔ `for` حساب کنید.

</div>

```python
total = 0
for i in range(2, 101, 2):
    total += i
print(total)
```

### تمرین ۲

با حلقهٔ `while` تعداد ارقام عدد `9876543` را بدون تبدیل به رشته بشمارید.

</div>

```python
n = 9876543
digits = 0
while n > 0:
    n //= 10
    digits += 1
print(digits)
```

### تمرین ۳

اعداد ۱ تا ۳۰ را چاپ کنید، ولی به جای مضرب‌های ۳ کلمهٔ `Fizz`، به جای مضرب‌های ۵ کلمهٔ `Buzz` و به جای مضرب‌های هر دو `FizzBuzz` چاپ کنید.

</div>

```python
for i in range(1, 31):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

### تمرین ۴

بررسی کنید عدد `97` اول است یا نه (عددی که جز ۱ و خودش مقسوم‌علیهی ندارد). از `break` استفاده کنید.

</div>

```python
n = 97
is_prime = n > 1
for d in range(2, int(n ** 0.5) + 1):
    if n % d == 0:
        is_prime = False
        break
print(n, "is prime" if is_prime else "is not prime")
```

### تمرین ۵

با حلقه‌های تودرتو یک مثلث وارونه از ستاره با ۵ سطر چاپ کنید (سطر اول ۵ ستاره و سطر آخر ۱ ستاره).

</div>

```python
for i in range(5, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```
