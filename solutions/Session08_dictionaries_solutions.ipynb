### تمرین ۱

دیکشنری یک کتاب با کلیدهای `title`، `author`، `year` و `price` بسازید. قیمت را ۱۰٪ افزایش دهید و کلید `pages` را اضافه کنید، سپس همهٔ اطلاعات را با حلقه چاپ کنید.

</div>

```python
book = {"title": "Python Basics", "author": "K. Mahmoudi", "year": 1405, "price": 350000}
book["price"] = book["price"] * 1.1
book["pages"] = 320
for key, value in book.items():
    print(f"{key}: {value}")
```

### تمرین ۲

تعداد تکرار هر حرف را در کلمهٔ `"mississippi"` با دیکشنری بشمارید.

</div>

```python
word = "mississippi"
counts = {}
for ch in word:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
```

### تمرین ۳

دیکشنری `stock = {"milk": 25, "bread": 0, "eggs": 120, "tea": 0}` را بگیرید و نام کالاهایی را که موجودی‌شان صفر است چاپ کنید.

</div>

```python
stock = {"milk": 25, "bread": 0, "eggs": 120, "tea": 0}
for item, qty in stock.items():
    if qty == 0:
        print(item)
```

### تمرین ۴

دو لیست `keys = ["a", "b", "c"]` و `values = [1, 2, 3]` را با `zip` به یک دیکشنری تبدیل کنید.

</div>

```python
keys = ["a", "b", "c"]
values = [1, 2, 3]
d = dict(zip(keys, values))
print(d)
```

### تمرین ۵

یک دفترچه تلفن به شکل لیستی از دیکشنری‌ها (با کلیدهای `name` و `phone`) برای ۳ نفر بسازید. سپس با حلقه شمارهٔ کسی را که نامش `"Sara"` است پیدا کنید و اگر نبود پیام مناسب چاپ کنید.

</div>

```python
contacts = [
    {"name": "Ali", "phone": "0911-111-1111"},
    {"name": "Sara", "phone": "0912-222-2222"},
    {"name": "Reza", "phone": "0935-333-3333"},
]
target = "Sara"
for c in contacts:
    if c["name"] == target:
        print(target, c["phone"])
        break
else:
    print(target, "not found")
```
