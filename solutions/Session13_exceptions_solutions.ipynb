### تمرین ۱

تابع `to_int(text)` بنویسید که رشته را به عدد صحیح تبدیل کند و اگر ممکن نبود `None` برگرداند. آن را روی `["12", "x", "7", "3.5"]` امتحان کنید.

</div>

```python
def to_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print([to_int(t) for t in ["12", "x", "7", "3.5"]])
```

### تمرین ۲

از لیست `["10", "20", "abc", "30", ""]` فقط مقدارهای قابل تبدیل به عدد را جمع کنید و تعداد مقدارهای نامعتبر را گزارش دهید.

</div>

```python
values = ["10", "20", "abc", "30", ""]
total = 0
invalid = 0
for v in values:
    try:
        total += int(v)
    except ValueError:
        invalid += 1
print("total:", total, "invalid:", invalid)
```

### تمرین ۳

تابع `get_item(lst, index)` بنویسید که اگر اندیس خارج از محدوده بود، به‌جای خطا پیام مناسب چاپ کند و `None` برگرداند.

</div>

```python
def get_item(lst, index):
    try:
        return lst[index]
    except IndexError:
        print(f"index {index} is out of range (size {len(lst)})")
        return None

print(get_item([1, 2, 3], 1))
print(get_item([1, 2, 3], 10))
```

### تمرین ۴

تابع `create_user(username, age)` بنویسید که اگر نام کاربری کمتر از ۳ کاراکتر بود یا سن بین ۱ و ۱۲۰ نبود، `ValueError` با پیام مناسب ایجاد کند؛ وگرنه دیکشنری کاربر را برگرداند.

</div>

```python
def create_user(username, age):
    if len(username) < 3:
        raise ValueError("username must have at least 3 characters")
    if not 1 <= age <= 120:
        raise ValueError("age must be between 1 and 120")
    return {"username": username, "age": age}

for args in [("keyvan", 39), ("ab", 20), ("sara", 150)]:
    try:
        print(create_user(*args))
    except ValueError as e:
        print("Error:", e)
```

### تمرین ۵

تابع `read_json(path)` بنویسید که فایل JSON را بخواند و اگر فایل وجود نداشت یا محتوایش JSON معتبر نبود (`json.JSONDecodeError`)، پیام مناسب چاپ کند و دیکشنری خالی برگرداند.

</div>

```python
import json

def read_json(path):
    try:
        with open(path, encoding="utf-8") as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"{path} not found")
    except json.JSONDecodeError as e:
        print(f"{path} is not valid JSON: {e}")
    return {}

with open("broken.json", "w", encoding="utf-8") as f:
    f.write("{not json")
print(read_json("missing.json"))
print(read_json("broken.json"))
```
