### تمرین ۱

کلاس `Rectangle` با ویژگی‌های `width` و `height` و متدهای `area()`، `perimeter()` و `is_square()` بنویسید.

</div>

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height


r = Rectangle(3, 4)
print(r.area(), r.perimeter(), r.is_square())
print(Rectangle(5, 5).is_square())
```

### تمرین ۲

کلاس `Counter` بسازید که متدهای `increment()`، `decrement()` (که کمتر از صفر نشود) و `reset()` داشته باشد و مقدارش با `print` نمایش داده شود.

</div>

```python
class Counter:
    def __init__(self):
        self.value = 0

    def increment(self):
        self.value += 1

    def decrement(self):
        if self.value > 0:
            self.value -= 1

    def reset(self):
        self.value = 0

    def __str__(self):
        return f"Counter({self.value})"


c = Counter()
c.increment()
c.increment()
c.decrement()
c.decrement()
c.decrement()
print(c)
```

### تمرین ۳

کلاس `ShoppingCart` بنویسید که کالاها را در یک دیکشنری `{name: (price, qty)}` نگه دارد و متدهای `add_item`، `remove_item` و `total` داشته باشد.

</div>

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, name, price, qty=1):
        if name in self.items:
            old_price, old_qty = self.items[name]
            self.items[name] = (price, old_qty + qty)
        else:
            self.items[name] = (price, qty)

    def remove_item(self, name):
        self.items.pop(name, None)

    def total(self):
        return sum(price * qty for price, qty in self.items.values())


cart = ShoppingCart()
cart.add_item("pen", 45000, 3)
cart.add_item("book", 280000)
cart.add_item("pen", 45000, 2)
cart.remove_item("book")
print(cart.items, cart.total())
```

### تمرین ۴

کلاس `Temperature` بسازید که دما را به سانتی‌گراد بگیرد و متدهای `to_fahrenheit()` و `to_kelvin()` داشته باشد. یک ویژگی کلاس `count` هم تعداد اشیای ساخته‌شده را بشمارد.

</div>

```python
class Temperature:
    count = 0

    def __init__(self, celsius):
        self.celsius = celsius
        Temperature.count += 1

    def to_fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    def to_kelvin(self):
        return self.celsius + 273.15


t1 = Temperature(25)
t2 = Temperature(-10)
print(t1.to_fahrenheit(), t2.to_kelvin(), Temperature.count)
```
