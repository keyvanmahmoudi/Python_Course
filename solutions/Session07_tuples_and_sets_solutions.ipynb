### تمرین ۲

تعداد کلمات **یکتا** در جملهٔ `"the cat and the dog and the bird"` را بشمارید.

</div>

```python
sentence = "the cat and the dog and the bird"
unique_words = set(sentence.split())
print(unique_words)
print(len(unique_words))
```

### تمرین ۳

دو لیست علاقه‌مندی `a = ["music", "books", "sport"]` و `b = ["sport", "movies", "music"]` داریم. علاقه‌مندی‌های مشترک و علاقه‌مندی‌هایی را که فقط در `a` هستند پیدا کنید.

</div>

```python
a = ["music", "books", "sport"]
b = ["sport", "movies", "music"]
print("common:", set(a) & set(b))
print("only a:", set(a) - set(b))
```

### تمرین ۵

لیستی از تاپل‌های `(name, score)` بسازید: `[("Ali", 17), ("Sara", 19), ("Reza", 14)]` و با حلقه و unpacking نام کسی را که بیشترین نمره را دارد پیدا کنید.

</div>

```python
results = [("Ali", 17), ("Sara", 19), ("Reza", 14)]
best_name, best_score = results[0]
for name, score in results:
    if score > best_score:
        best_name, best_score = name, score
print(best_name, best_score)
```
