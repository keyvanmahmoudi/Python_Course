### تمرین ۱

رشتهٔ `"Artificial Intelligence"` را در متغیر بگذارید و اولین کاراکتر، آخرین کاراکتر، کلمهٔ اول و نسخهٔ برعکس آن را چاپ کنید.

</div>

```python
s = "Artificial Intelligence"
print(s[0])
print(s[-1])
print(s.split()[0])
print(s[::-1])
```

### تمرین ۲

در جملهٔ `"I love Java. Java is great."` همهٔ `Java`ها را با `Python` جایگزین کنید و تعداد کلمات جمله را چاپ کنید.

</div>

```python
sentence = "I love Java. Java is great."
new_sentence = sentence.replace("Java", "Python")
print(new_sentence)
print("Words:", len(new_sentence.split()))
```
