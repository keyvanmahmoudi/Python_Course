### تمرین ۱

سه عدد در متغیرهای `a`، `b` و `c` بگذارید و بدون استفاده از تابع `max` بزرگ‌ترین آن‌ها را چاپ کنید.

</div>

```python
a, b, c = 12, 45, 27
if a >= b and a >= c:
    largest = a
elif b >= c:
    largest = b
else:
    largest = c
print("Largest:", largest)
```

### تمرین ۲

سالی را در متغیر `year` بگذارید و بررسی کنید کبیسهٔ میلادی است یا نه (بخش‌پذیر بر ۴ و نه بر ۱۰۰، یا بخش‌پذیر بر ۴۰۰).

</div>

```python
year = 2024
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

### تمرین ۳

برنامه‌ای بنویسید که سن را بگیرد و گروه سنی را چاپ کند: کمتر از ۱۲ «کودک»، ۱۲ تا ۱۷ «نوجوان»، ۱۸ تا ۵۹ «بزرگسال» و ۶۰ به بالا «سالمند».

</div>

```python
age = 35
if age < 12:
    group = "child"
elif age < 18:
    group = "teenager"
elif age < 60:
    group = "adult"
else:
    group = "senior"
print(group)
```

### تمرین ۴

رمز عبوری را در متغیر بگذارید و بررسی کنید معتبر است یا نه: دست‌کم ۸ کاراکتر داشته باشد و فقط از رقم تشکیل نشده باشد.

</div>

```python
password = "abc12345"
if len(password) >= 8 and not password.isdigit():
    print("Valid password")
else:
    print("Invalid password")
```

### تمرین ۵

با عبارت شرطی یک‌خطی، قدر مطلق عدد `n = -15` را بدون تابع `abs` حساب و چاپ کنید.

</div>

```python
n = -15
absolute = n if n >= 0 else -n
print(absolute)
```
