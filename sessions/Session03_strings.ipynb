## ۱. ساخت رشته

رشته دنباله‌ای از کاراکترهاست که بین `'...'` یا `"..."` قرار می‌گیرد. هر دو یکسان‌اند؛ اگر داخل متن
کوتیشن داریم، از نوع دیگر برای دور آن استفاده می‌کنیم. رشته‌های چندخطی را بین سه کوتیشن (سه `"` یا سه `'`) می‌نویسیم.

</div>

```python
s1 = 'Hello'
s2 = "It's a nice day"
s3 = """This is
a multi-line
string"""
print(s1)
print(s2)
print(s3)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

**کاراکترهای ویژه (Escape):** با `\` شروع می‌شوند. `\n` به خط بعد می‌رود و `\t` یک فاصلهٔ تب می‌گذارد.

</div>

```python
print("Name:\tAli\nAge:\t21")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

عملگر `+` دو رشته را به هم می‌چسباند و `*` یک رشته را تکرار می‌کند. تابع `len` طول رشته را می‌دهد:

</div>

```python
first = "Keyvan"
last = "Mahmoudi"
full = first + " " + last
print(full)
print("=" * len(full))
print(len(full))
```

## ۲. اندیس و برش (Indexing & Slicing)

هر کاراکتر رشته یک **اندیس** (شماره) دارد که **از صفر شروع می‌شود**. اندیس‌های منفی از انتها می‌شمارند:
`-1` آخرین کاراکتر است.

| کاراکتر | `P` | `y` | `t` | `h` | `o` | `n` |
|---|---|---|---|---|---|---|
| اندیس مثبت | 0 | 1 | 2 | 3 | 4 | 5 |
| اندیس منفی | -6 | -5 | -4 | -3 | -2 | -1 |

با **برش** `s[start:stop:step]` بخشی از رشته را جدا می‌کنیم. `start` شامل می‌شود ولی `stop` **شامل نمی‌شود**.
هر کدام را خالی بگذاریم، مقدار پیش‌فرض (ابتدا، انتها، گام ۱) استفاده می‌شود.

</div>

```python
s = "Python"
print(s[0])
print(s[5])
print(s[-1])
print(s[-2])
```

```python
s = "Python Programming"
print(s[0:6])
print(s[7:])
print(s[:6])
print(s[::2])
print(s[::-1])
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

اگر اندیسی خارج از محدوده بخواهیم، خطای `IndexError` می‌گیریم (اما برش هرگز خطا نمی‌دهد):

</div>

```python
s = "Python"
print(s[2:100])
print(s[10])
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۳. تغییرناپذیری رشته‌ها

رشته‌ها **تغییرناپذیر (immutable)** هستند: نمی‌توان یک کاراکترشان را مستقیم عوض کرد.
برای تغییر، باید یک رشتهٔ جدید بسازیم.

</div>

```python
s = "Python"
s[0] = "J"
```

```python
s = "Python"
new_s = "J" + s[1:]
print(new_s)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۴. متدهای پرکاربرد رشته

**متد** تابعی است که به یک نوع داده تعلق دارد و با نقطه صدا زده می‌شود: `text.upper()`.
متدهای رشته، رشتهٔ اصلی را تغییر نمی‌دهند و **رشتهٔ جدیدی برمی‌گردانند**.

| متد | کار |
|---|---|
| `upper()` / `lower()` / `title()` | تبدیل به حروف بزرگ / کوچک / عنوانی |
| `strip()` | حذف فاصله‌های ابتدا و انتها |
| `replace(a, b)` | جایگزینی `a` با `b` |
| `split(sep)` | شکستن رشته به لیستی از تکه‌ها |
| `sep.join(list)` | چسباندن تکه‌ها با جداکننده |
| `find(x)` / `count(x)` | محل اولین وقوع / تعداد وقوع |
| `startswith(x)` / `endswith(x)` | آیا با `x` شروع / تمام می‌شود؟ |
| `isdigit()` | آیا فقط رقم است؟ |

</div>

```python
text = "  Python is Easy and Python is Fun  "
print(text.upper())
print(text.lower())
print(text.strip())
print(text.replace("Python", "Java"))
print(text.count("Python"))
print(text.find("Easy"))
```

```python
csv_line = "Ali,21,Amol,Computer"
parts = csv_line.split(",")
print(parts)

print(" | ".join(parts))
```

```python
email = "keyvan@example.com"
print(email.endswith(".com"))
print("12345".isdigit(), "12a45".isdigit())
```

```python
sentence = "Machine learning is part of AI"
print("learning" in sentence)
print("Deep" in sentence)
```

## ۵. قالب‌بندی با f-string

روش مدرن و توصیه‌شده برای ساختن متن از متغیرها، **f-string** است: قبل از کوتیشن حرف `f` می‌گذاریم و
هر عبارت پایتون را داخل `{}` می‌نویسیم.

بعد از `:` داخل آکولاد می‌توانیم **قالب** نمایش را تعیین کنیم:

| قالب | معنی | نمونه |
|---|---|---|
| `{x:.2f}` | دو رقم اعشار | `3.14` |
| `{x:,}` | جداکنندهٔ هزارگان | `12,500,000` |
| `{x:.1%}` | درصد | `85.0%` |
| `{s:>10}` / `{s:<10}` / `{s:^10}` | تراز راست / چپ / وسط در ۱۰ کاراکتر | |

</div>

```python
name = "Sara"
score = 18.4567
print(f"{name}'s score is {score}")
print(f"{name}'s score is {score:.2f}")
print(f"Next year {name} will be {21 + 1} years old")
```

```python
salary = 32500000
ratio = 0.8534
print(f"Salary: {salary:,} Toman")
print(f"Success rate: {ratio:.1%}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

با تراز کردن می‌توانیم خروجی جدول‌مانند بسازیم:

</div>

```python
print(f"{'Item':<10}{'Qty':>5}{'Price':>12}")
print(f"{'Pen':<10}{3:>5}{45000:>12,}")
print(f"{'Notebook':<10}{2:>5}{180000:>12,}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

> 💡 **نکته:** در کدهای قدیمی‌تر روش‌های `"%d" % x` و `"{}".format(x)` را هم می‌بینید. کار می‌کنند، ولی f-string خواناتر و سریع‌تر است.

</div>
