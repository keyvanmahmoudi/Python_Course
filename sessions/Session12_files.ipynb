## ۱. باز کردن فایل با with

تابع `open(path, mode, encoding)` فایل را باز می‌کند. روش درست، استفاده از دستور `with` است که فایل را در پایان
بلوک **خودکار می‌بندد**، حتی اگر خطایی رخ دهد.

| حالت | کار |
|---|---|
| `"r"` | خواندن (پیش‌فرض)؛ اگر فایل نباشد خطا |
| `"w"` | نوشتن؛ اگر فایل باشد **محتوایش پاک می‌شود** |
| `"a"` | افزودن به انتهای فایل |
| `"x"` | ساخت فایل جدید؛ اگر باشد خطا |

> برای متن فارسی همیشه `encoding="utf-8"` را بنویسید؛ وگرنه روی ویندوز ممکن است متن خراب شود.

</div>

```python
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("سلام دنیا\n")
    f.write("Python is fun\n")
    f.write("Files are easy\n")
print("file written")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۲. خواندن فایل

سه روش اصلی خواندن:

- `f.read()`: کل فایل به شکل یک رشته
- `f.readlines()`: لیستی از خطوط
- `for line in f`: خط‌به‌خط؛ **بهترین روش برای فایل‌های بزرگ** چون کل فایل یک‌جا در حافظه بارگذاری نمی‌شود

هر خط در انتها `\n` دارد که معمولاً با `strip()` حذفش می‌کنیم.

</div>

```python
with open("notes.txt", encoding="utf-8") as f:
    content = f.read()
print(repr(content))
```

```python
with open("notes.txt", encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        print(number, line.strip())
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۳. افزودن به فایل

حالت `"a"` محتوای قبلی را نگه می‌دارد و متن جدید را به انتها اضافه می‌کند؛ مناسب برای فایل‌های گزارش (log).
`print` هم با پارامتر `file` می‌تواند در فایل بنویسد.

</div>

```python
from datetime import datetime

for event in ["start", "load data", "finish"]:
    with open("log.txt", "a", encoding="utf-8") as f:
        print(f"{datetime(2026, 11, 8, 10, 0):%H:%M} {event}", file=f)

with open("log.txt", encoding="utf-8") as f:
    print(f.read())
```

## ۴. فایل CSV

**CSV** (مقادیر جداشده با ویرگول) ساده‌ترین قالب دادهٔ جدولی است: هر خط یک رکورد و مقدارها با ویرگول جدا شده‌اند.
خط اول معمولاً **سرستون‌ها** است. این قالب را Excel و pandas به‌راحتی می‌خوانند.

ماژول استاندارد `csv` جزئیاتی مثل ویرگول داخل مقدار را درست مدیریت می‌کند؛ پس به‌جای `split(",")` از آن استفاده کنید:

- `csv.writer` و `csv.reader`: کار با هر سطر به شکل لیست
- `csv.DictWriter` و `csv.DictReader`: کار با هر سطر به شکل دیکشنری (خواناتر)

> هنگام باز کردن فایل CSV برای نوشتن، `newline=""` را بگذارید تا روی ویندوز خط خالی اضافه ایجاد نشود.

</div>

```python
import csv

students = [
    {"name": "Ali", "city": "Amol", "score": 17.5},
    {"name": "Sara", "city": "Sari", "score": 19},
    {"name": "Reza", "city": "Babol", "score": 14.25},
    {"name": "Mina", "city": "Amol", "score": 16},
]

with open("students.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "city", "score"])
    writer.writeheader()
    writer.writerows(students)

with open("students.csv", encoding="utf-8") as f:
    print(f.read())
```

```python
import csv

with open("students.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

print(rows[0])
scores = [float(r["score"]) for r in rows]
print(f"average: {sum(scores) / len(scores):.2f}")

amol = [r["name"] for r in rows if r["city"] == "Amol"]
print("from Amol:", amol)
```

## ۵. فایل JSON

**JSON** رایج‌ترین قالب تبادل داده در وب و APIهاست و ساختاری تقریباً یکسان با دیکشنری و لیست پایتون دارد.
ماژول `json` این تبدیل را انجام می‌دهد:

| تابع | کار |
|---|---|
| `json.dump(obj, f)` | نوشتن شیء پایتون در فایل |
| `json.load(f)` | خواندن فایل به شیء پایتون |
| `json.dumps(obj)` | تبدیل به رشتهٔ JSON |
| `json.loads(text)` | تبدیل رشتهٔ JSON به شیء پایتون |

برای حفظ حروف فارسی از `ensure_ascii=False` و برای خوانایی از `indent=2` استفاده کنید.

</div>

```python
import json

library = {
    "name": "کتابخانهٔ آموزشگاه",
    "books": [
        {"title": "Python Crash Course", "year": 2023, "available": True},
        {"title": "Fluent Python", "year": 2022, "available": False},
    ],
}

with open("library.json", "w", encoding="utf-8") as f:
    json.dump(library, f, ensure_ascii=False, indent=2)

with open("library.json", encoding="utf-8") as f:
    print(f.read())
```

```python
import json

with open("library.json", encoding="utf-8") as f:
    data = json.load(f)

print(type(data))
for book in data["books"]:
    status = "available" if book["available"] else "on loan"
    print(f"{book['title']} ({book['year']}): {status}")
```

```python
text = '{"user": "keyvan", "scores": [18, 19.5]}'
obj = json.loads(text)
print(obj["scores"][1])
print(json.dumps(obj))
```

## ۶. مسیرها با pathlib

ماژول `pathlib` روش مدرن و مستقل از سیستم‌عامل برای کار با مسیر فایل‌ها و پوشه‌هاست.
عملگر `/` مسیرها را به هم می‌چسباند و متدهایی مثل `exists`، `mkdir`، `glob`، `read_text` و `write_text` دارد.

</div>

```python
from pathlib import Path

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

for month in ["farvardin", "ordibehesht", "khordad"]:
    (data_dir / f"{month}.txt").write_text(f"report for {month}", encoding="utf-8")

for path in sorted(data_dir.glob("*.txt")):
    print(path.name, path.stem, path.suffix, "->", path.read_text(encoding="utf-8"))
```
