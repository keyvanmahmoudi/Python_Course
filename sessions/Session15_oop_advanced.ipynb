## ۱. وراثت

**وراثت (inheritance)** یعنی ساختن کلاس جدید (**فرزند**) بر پایهٔ کلاس موجود (**والد**). فرزند همهٔ ویژگی‌ها و متدهای
والد را به ارث می‌برد و می‌تواند ویژگی‌ها و متدهای جدید اضافه کند یا متدهای والد را تغییر دهد.

رابطهٔ وراثت باید **«هست یک» (is-a)** باشد: دانشجو **یک** شخص است؛ حساب پس‌انداز **یک** حساب بانکی است.

`super()` به کلاس والد اشاره دارد؛ معمولاً در `__init__` فرزند، سازندهٔ والد را با آن صدا می‌زنیم.

</div>

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"I am {self.name}, {self.age} years old."


class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age)
        self.major = major

    def study(self):
        return f"{self.name} is studying {self.major}."


s = Student("Sara", 21, "Computer Engineering")
print(s.introduce())
print(s.study())
print(isinstance(s, Student), isinstance(s, Person))
```

## ۲. بازنویسی متد و چندریختی

فرزند می‌تواند متدی هم‌نام با متد والد تعریف کند تا رفتار آن را **بازنویسی (override)** کند. با `super().method()`
می‌توان نسخهٔ والد را هم صدا زد و آن را گسترش داد.

**چندریختی (polymorphism)** یعنی اشیای کلاس‌های مختلف به یک متد هم‌نام پاسخ می‌دهند، هر کدام به روش خودش.
در نتیجه می‌توانیم با یک حلقه روی اشیای مختلف کار کنیم بدون اینکه نوعشان برایمان مهم باشد.

</div>

```python
import math


class Shape:
    def area(self):
        raise NotImplementedError("subclasses must implement area()")

    def describe(self):
        return f"{type(self).__name__} with area {self.area():.2f}"


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)


shapes = [Circle(1), Rectangle(3, 4), Square(2)]
for shape in shapes:
    print(shape.describe())

print("total area:", round(sum(sh.area() for sh in shapes), 2))
```

مثال گسترش متد والد با `super()`:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am {self.name}."


class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def introduce(self):
        base = super().introduce()
        return f"{base} I teach {self.subject}."


print(Teacher("Keyvan", "Python").introduce())
```

## ۳. متدهای ویژه

با تعریف متدهای ویژه، اشیای ما می‌توانند با توابع و عملگرهای داخلی پایتون کار کنند:

| متد | فراخوانی |
|---|---|
| `__str__` | `print(obj)`، `str(obj)`: نمایش برای کاربر |
| `__repr__` | نمایش در کنسول و برای برنامه‌نویس |
| `__len__` | `len(obj)` |
| `__eq__` | `obj1 == obj2` |
| `__lt__` | `obj1 < obj2` (و در نتیجه `sorted`) |
| `__add__` | `obj1 + obj2` |
| `__getitem__` | `obj[i]` |

</div>

```python
class Vector:
    """A 2D vector."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, k):
        return Vector(self.x * k, self.y * k)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __abs__(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5


v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)
print(v1 * 3)
print(abs(v2))
print(v1 == Vector(1, 2))
v1
```

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, index):
        return self.songs[index]

    def __str__(self):
        return f"Playlist '{self.name}' with {len(self)} songs"


p = Playlist("Study")
for song in ["Song A", "Song B", "Song C"]:
    p.add(song)

print(p)
print(len(p))
print(p[0], p[-1])
for song in p:
    print("-", song)
```

## ۴. کپسوله‌سازی و property

**کپسوله‌سازی (encapsulation)** یعنی محافظت از دادهٔ درونی شیء تا فقط از راه‌های کنترل‌شده تغییر کند.

- پایتون «خصوصی» واقعی ندارد؛ **قرارداد** این است که نامی که با `_` شروع می‌شود (مثل `_balance`) داخلی است
  و نباید از بیرون مستقیم تغییر کند.
- دکوراتور `@property` اجازه می‌دهد یک متد مثل ویژگی خوانده شود و با `@<name>.setter` هنگام تغییر،
  مقدار جدید **اعتبارسنجی** شود.

</div>

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score          # goes through the setter below

    @property
    def score(self):
        return self._score

    @score.setter
    def score(self, value):
        if not 0 <= value <= 20:
            raise ValueError(f"score must be between 0 and 20, got {value}")
        self._score = value

    @property
    def passed(self):
        return self.score >= 10


s = Student("Reza", 14)
print(s.score, s.passed)

s.score = 8
print(s.score, s.passed)

try:
    s.score = 25
except ValueError as e:
    print("Error:", e)
```

## ۵. ترکیب در برابر وراثت

همه‌جا وراثت بهترین انتخاب نیست. اگر رابطه **«دارد یک» (has-a)** است، از **ترکیب (composition)** استفاده کنید:
یعنی شیء کلاس دیگر را به‌عنوان ویژگی نگه دارید. مثلاً کلاس `Course` **دارای** لیستی از `Student`هاست، نه اینکه خودش
دانشجو باشد.

</div>

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __repr__(self):
        return f"Student({self.name!r}, {self.score})"


class Course:
    def __init__(self, title):
        self.title = title
        self.students = []

    def enroll(self, student):
        self.students.append(student)

    def average(self):
        return sum(s.score for s in self.students) / len(self.students)

    def top(self, n=2):
        return sorted(self.students, key=lambda s: s.score, reverse=True)[:n]

    def __len__(self):
        return len(self.students)


course = Course("Python")
for name, score in [("Ali", 17), ("Sara", 19), ("Reza", 14), ("Mina", 18)]:
    course.enroll(Student(name, score))

print(len(course), "students")
print(f"average: {course.average():.2f}")
print("top:", course.top())
```
