## ۱. متغیر چیست؟

**متغیر** نامی است که به یک مقدار در حافظه اشاره می‌کند. مثل برچسبی که روی یک جعبه می‌چسبانیم تا بعداً
بتوانیم محتوای جعبه را پیدا کنیم. با علامت `=` (عملگر **انتساب**) مقداری را به یک نام نسبت می‌دهیم.

> علامت `=` در برنامه‌نویسی یعنی «مقدار سمت راست را در نام سمت چپ قرار بده»، نه «برابر است با».

</div>

```python
name = "Sara"
age = 21
height = 1.68
print(name, age, height)
```

<div dir="rtl">

مقدار متغیر قابل تغییر است. در خط دوم زیر، ابتدا سمت راست (`score + 5`) محاسبه و سپس در `score` ذخیره می‌شود:

</div>

```python
score = 10
score = score + 5
print(score)
```

<div dir="rtl">

می‌توانیم در یک خط به چند متغیر مقدار بدهیم، یا مقدار دو متغیر را با هم جابه‌جا کنیم:

</div>

```python
x, y = 3, 7
print(x, y)
x, y = y, x   # swap
print(x, y)
```

## ۳. انواع دادهٔ پایه

هر مقدار در پایتون یک **نوع (type)** دارد. چهار نوع پایه:

| نوع | نام | مثال |
|---|---|---|
| `int` | عدد صحیح | `42`، `-7`، `0` |
| `float` | عدد اعشاری | `3.14`، `-0.5`، `2.0` |
| `str` | رشته (متن) | `"سلام"`، `'Python'` |
| `bool` | منطقی (درست/نادرست) | `True`، `False` |

با تابع `type` نوع هر مقدار را می‌بینیم:

</div>

```python
print(type(42))
print(type(3.14))
print(type("Python"))
print(type(True))
```

<div dir="rtl">

برای اعداد بزرگ می‌توانیم برای خوانایی از `_` بین رقم‌ها استفاده کنیم. همچنین مقدار ویژهٔ `None` به معنای «هیچ مقداری»
است و نوع خودش (`NoneType`) را دارد:

</div>

```python
population = 85_000_000
print(population)

result = None
print(result, type(result))
```

## ۴. عملگرهای حسابی

| عملگر | کار | مثال | نتیجه |
|---|---|---|---|
| `+` | جمع | `7 + 2` | `9` |
| `-` | تفریق | `7 - 2` | `5` |
| `*` | ضرب | `7 * 2` | `14` |
| `/` | تقسیم (همیشه float) | `7 / 2` | `3.5` |
| `//` | تقسیم صحیح | `7 // 2` | `3` |
| `%` | باقی‌مانده | `7 % 2` | `1` |
| `**` | توان | `7 ** 2` | `49` |

</div>

```python
a = 17
b = 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
```

<div dir="rtl">

**ترتیب اجرای عملگرها** مثل ریاضی است: ابتدا پرانتز، سپس توان، بعد ضرب و تقسیم، و در آخر جمع و تفریق.
هر جا شک داشتید، از پرانتز استفاده کنید؛ کد خواناتر هم می‌شود.

</div>

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3 ** 2)  # power is right-associative: 2 ** 9
```

<div dir="rtl">

> 💡 **نکته:** عملگر `%` کاربردهای زیادی دارد: `n % 2 == 0` یعنی `n` زوج است، و `n % 10` رقم یکان عدد را می‌دهد.

</div>

```python
n = 1405
print("last digit:", n % 10)
print("is even?", n % 2 == 0)
```

## ۵. عملگرهای انتساب ترکیبی

وقتی می‌خواهیم مقدار یک متغیر را بر اساس مقدار فعلی‌اش تغییر دهیم، شکل کوتاه‌تری داریم:
`x += 5` معادل `x = x + 5` است. همین الگو برای `-=`، `*=`، `/=`، `//=`، `%=` و `**=` هم وجود دارد.

</div>

```python
balance = 1000
balance += 250
balance -= 100
balance *= 2
print(balance)
```

## ۶. تبدیل نوع داده

گاهی باید نوع یک مقدار را عوض کنیم. توابع `int()`، `float()`، `str()` و `bool()` این کار را انجام می‌دهند.

</div>

```python
print(int("25") + 1)
print(float("2.5") * 2)
print(int(3.9))
print(round(3.9), round(3.14159, 2))
print(str(100) + "0")
```

<div dir="rtl">

جمع مستقیم رشته و عدد خطای `TypeError` می‌دهد، چون پایتون نمی‌داند منظور ما جمع ریاضی است یا چسباندن متن:

</div>

```python
age = 25
print("Age: " + age)
```

<div dir="rtl">

راه درست، تبدیل عدد به رشته یا جدا کردن با ویرگول در `print` است:

</div>

```python
print("Age: " + str(age))
print("Age:", age)
```

<div dir="rtl">

تابع `bool` هر مقداری را به `True` یا `False` تبدیل می‌کند. عدد صفر و رشتهٔ خالی `False` و بقیه `True` هستند:

</div>

```python
print(bool(0), bool(7), bool(""), bool("Hi"))
```

## ۷. گرفتن ورودی از کاربر

تابع `input` متنی را به کاربر نشان می‌دهد، منتظر می‌ماند تا کاربر چیزی تایپ کند و Enter بزند، و آن را برمی‌گرداند.

**مهم:** خروجی `input` **همیشه رشته** است؛ حتی اگر کاربر عدد وارد کند. برای محاسبه باید آن را با `int` یا `float` تبدیل کنیم.

</div>

```python
name = input("Your name: ")
birth_year = int(input("Your birth year (Solar Hijri): "))
print("Hi", name)
print("You are about", 1405 - birth_year, "years old.")
```

<div dir="rtl">

ببینید اگر ورودی را تبدیل نکنیم چه می‌شود: دو رشته کنار هم چسبانده می‌شوند، نه جمع:

</div>

```python
a = input("First number: ")
b = input("Second number: ")
print("Without conversion:", a + b)
print("With conversion:", int(a) + int(b))
```

## ۸. پروژهٔ کوچک: محاسبهٔ قیمت با تخفیف

برنامه‌ای می‌نویسیم که قیمت کالا و درصد تخفیف را از کاربر بگیرد و مبلغ تخفیف و قیمت نهایی را حساب کند.

</div>

```python
price = float(input("Price (Toman): "))
discount_percent = float(input("Discount (%): "))

discount = price * discount_percent / 100
final_price = price - discount

print("Discount amount:", discount)
print("Final price:", final_price)
```
