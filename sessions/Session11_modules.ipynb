## ۱. ماژول و import

**ماژول** یک فایل پایتون (`.py`) است که توابع، کلاس‌ها و متغیرهایی دارد. **بسته (package)** مجموعه‌ای از ماژول‌هاست.
برای استفاده از ماژول باید آن را `import` کنیم:

| روش | استفاده |
|---|---|
| `import math` | `math.sqrt(16)` |
| `import numpy as np` | با نام کوتاه: `np.array(...)` |
| `from math import sqrt, pi` | مستقیم: `sqrt(16)` |

> از `from module import *` استفاده نکنید: معلوم نیست هر نام از کجا آمده و ممکن است نام‌های شما را بازنویسی کند.

</div>

```python
import math
print(math.sqrt(16))

from math import pi, floor
print(pi, floor(3.7))

import datetime as dt
print(dt.date(2026, 11, 8))
```

## ۲. math: توابع ریاضی

ماژول `math` توابع و ثابت‌های ریاضی را دارد.

</div>

```python
import math
print("sqrt:", math.sqrt(2))
print("ceil/floor:", math.ceil(4.1), math.floor(4.9))
print("log:", math.log(100, 10), math.log(math.e))
print("sin 90:", math.sin(math.radians(90)))
print("pi, e:", math.pi, math.e)
```

## ۳. random: اعداد تصادفی

ماژول `random` برای شبیه‌سازی، بازی و نمونه‌گیری به کار می‌رود. با `random.seed` نتایج تکرارپذیر می‌شوند؛
این در علم داده و یادگیری ماشین بسیار مهم است تا آزمایش‌ها قابل تکرار باشند.

</div>

```python
import random

random.seed(2026)
print(random.randint(1, 6))
print(random.random())
print(random.choice(["rock", "paper", "scissors"]))
print(random.sample(range(1, 50), 6))

cards = ["A", "K", "Q", "J"]
random.shuffle(cards)
print(cards)
```

## ۴. datetime: تاریخ و زمان

ماژول `datetime` با تاریخ و زمان (میلادی) کار می‌کند: تاریخ امروز، اختلاف دو تاریخ و قالب‌بندی.
برای تاریخ شمسی کتابخانه‌های بیرونی مثل `jdatetime` وجود دارند که با `pip` نصب می‌شوند.

</div>

```python
from datetime import date, datetime, timedelta

start = date(2026, 11, 8)
exam = date(2026, 12, 20)
print("days until exam:", (exam - start).days)
print("one week after exam:", exam + timedelta(weeks=1))
print(exam.strftime("%A, %d %B %Y"))

moment = datetime(2026, 11, 8, 20, 30)
print(moment.strftime("%Y-%m-%d %H:%M"))
```

## ۵. collections و os

- `collections.Counter`: شمارش سریع تکرارها؛ همان کاری که در جلسهٔ ۸ با دیکشنری کردیم.
- `collections.defaultdict`: دیکشنری‌ای که برای کلید ناموجود خودکار مقدار پیش‌فرض می‌سازد.
- `os`: کار با سیستم‌عامل، مسیرها و پوشه‌ها.

</div>

```python
from collections import Counter, defaultdict

text = "to be or not to be that is the question to ask"
counts = Counter(text.split())
print(counts)
print(counts.most_common(3))

groups = defaultdict(list)
for name, city in [("Ali", "Amol"), ("Sara", "Sari"), ("Reza", "Amol")]:
    groups[city].append(name)
print(dict(groups))
```

## ۶. ساخت ماژول شخصی

هر فایل `.py` که بنویسید یک ماژول است. کافی است فایل کنار نوت‌بوک (یا برنامه) باشد تا بتوانید آن را import کنید.
در Jupyter با دستور جادویی `%%writefile` می‌توان محتوای یک سلول را در فایل نوشت.

شرط `if __name__ == "__main__":` بخشی از کد را فقط وقتی اجرا می‌کند که فایل **مستقیم** اجرا شود، نه وقتی import می‌شود.

</div>

```python
%%writefile geometry.py
"""Small geometry helpers for the Python course."""
import math


def circle_area(radius):
    """Return the area of a circle."""
    return math.pi * radius ** 2


def rectangle_area(width, height):
    """Return the area of a rectangle."""
    return width * height


if __name__ == "__main__":
    print(circle_area(1))
```

```python
import geometry

print(geometry.circle_area(2))
print(geometry.rectangle_area(3, 4))
print(geometry.__doc__)
```

## ۷. pip و محیط مجازی

کتابخانه‌های بیرونی (مثل `numpy`، `pandas` و `requests`) از مخزن **PyPI** با ابزار `pip` نصب می‌شوند.

| دستور (در ترمینال) | کار |
|---|---|
| `pip install requests` | نصب |
| `pip install pandas==2.2.0` | نصب نسخهٔ مشخص |
| `pip install -U requests` | به‌روزرسانی |
| `pip uninstall requests` | حذف |
| `pip list` | فهرست کتابخانه‌های نصب‌شده |
| `pip freeze > requirements.txt` | ذخیرهٔ فهرست وابستگی‌ها |
| `pip install -r requirements.txt` | نصب همهٔ وابستگی‌ها |

در نوت‌بوک Jupyter یا Colab از `%pip install requests` استفاده کنید.

**محیط مجازی (virtual environment):** هر پروژه بهتر است محیط جداگانهٔ خودش را داشته باشد تا نسخه‌های کتابخانه‌ها
بین پروژه‌ها تداخل نکنند:

| مرحله | ویندوز | لینوکس / مک |
|---|---|---|
| ساخت | `python -m venv .venv` | `python3 -m venv .venv` |
| فعال‌سازی | `.venv\Scripts\activate` | `source .venv/bin/activate` |
| غیرفعال‌سازی | `deactivate` | `deactivate` |

کاربران Anaconda می‌توانند از `conda create -n myenv python=3.12` و `conda activate myenv` استفاده کنند.

</div>
