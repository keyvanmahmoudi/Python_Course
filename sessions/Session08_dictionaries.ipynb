# جلسهٔ ۸: دیکشنری

**دیکشنری** داده‌ها را به شکل جفت‌های **کلید: مقدار** نگه می‌دارد؛ درست مثل یک لغت‌نامه که با داشتن کلمه (کلید)
معنی (مقدار) را پیدا می‌کنیم. دیکشنری یکی از مهم‌ترین ساختارهای پایتون است و پایهٔ فرمت JSON و
ساختار جدول‌های pandas در ترم بعد است.

## ۱. ساخت دیکشنری

دیکشنری با آکولاد ساخته می‌شود و هر عضو آن یک جفت `key: value` است.

- **کلیدها یکتا هستند** و باید تغییرناپذیر باشند (معمولاً رشته یا عدد، یا تاپل).
- مقدارها می‌توانند هر چیزی باشند: عدد، رشته، لیست یا حتی دیکشنری دیگر.
- از پایتون ۳.۷ به بعد، ترتیب درج کلیدها حفظ می‌شود.

</div>

```python
student = {
    "name": "Sara",
    "age": 21,
    "major": "Computer Engineering",
    "courses": ["Python", "SQL"],
}
print(student)
print(student["name"])
print(student["courses"][0])
print(len(student))
```

<div dir="rtl">

اگر کلیدی را بخواهیم که وجود ندارد، خطای `KeyError` می‌گیریم:

</div>

```python
student["gpa"]
```

<div dir="rtl">

## ۲. دسترسی امن با get

متد `get(key, default)` اگر کلید وجود داشته باشد مقدارش را برمی‌گرداند و اگر نباشد، مقدار پیش‌فرض
(یا `None`) را؛ بدون خطا. برای بررسی وجود کلید هم از `in` استفاده می‌کنیم.

</div>

```python
print(student.get("gpa"))
print(student.get("gpa", "not set"))
print("age" in student, "gpa" in student)
```

<div dir="rtl">

## ۳. افزودن، تغییر و حذف

| عملیات | کار |
|---|---|
| `d[key] = value` | اگر کلید باشد تغییر، وگرنه افزودن |
| `d.update(other)` | افزودن یا تغییر چند جفت با هم |
| `d.pop(key)` | حذف و برگرداندن مقدار |
| `del d[key]` | حذف |

</div>

```python
stock = {"milk": 25, "bread": 40}
stock["eggs"] = 120
stock["milk"] = stock["milk"] - 5
print(stock)

stock.update({"tea": 10, "bread": 35})
print(stock)

removed = stock.pop("bread")
print("removed:", removed, "->", stock)
```

## ۴. پیمایش دیکشنری

سه متد برای پیمایش داریم:

- `keys()`: کلیدها
- `values()`: مقدارها
- `items()`: جفت‌های `(کلید، مقدار)` به شکل تاپل؛ **پرکاربردترین**

حلقهٔ `for key in d` به‌طور پیش‌فرض روی کلیدها حرکت می‌کند.

</div>

```python
prices = {"pen": 45000, "book": 280000, "bag": 950000}

print(list(prices.keys()))
print(list(prices.values()))
print("total:", sum(prices.values()))

for item, price in prices.items():
    print(f"{item:<6}{price:>10,}")
```

<div dir="rtl">

پیدا کردن گران‌ترین کالا: تابع `max` با پارامتر `key` می‌تواند کلیدی را برگرداند که بیشترین مقدار را دارد:

</div>

```python
most_expensive = max(prices, key=prices.get)
print(most_expensive, prices[most_expensive])
```

## ۵. ساختارهای تودرتو

در دنیای واقعی داده‌ها اغلب به شکل **لیستی از دیکشنری‌ها** هستند؛ هر دیکشنری یک «رکورد» (مثل یک سطر جدول) است.
این دقیقاً همان شکلی است که دادهٔ JSON از وب‌سرویس‌ها می‌آید و pandas آن را به جدول تبدیل می‌کند.

</div>

```python
students = [
    {"name": "Ali", "city": "Amol", "score": 17},
    {"name": "Sara", "city": "Sari", "score": 19},
    {"name": "Reza", "city": "Amol", "score": 14},
]

for st in students:
    print(f"{st['name']:<6}{st['city']:<6}{st['score']}")

amol_students = []
for st in students:
    if st["city"] == "Amol":
        amol_students.append(st["name"])
print("from Amol:", amol_students)
```

```python
grades = {
    "Ali": [18, 17, 19],
    "Sara": [15, 16, 14],
}
grades["Reza"] = [20, 19, 18]

for name, marks in grades.items():
    print(f"{name}: average = {sum(marks) / len(marks):.2f}")
```

## ۶. شمارش با دیکشنری

یک الگوی بسیار رایج: شمردن تعداد تکرار هر چیز. کلید = آن چیز، مقدار = تعداد.
با `d.get(key, 0) + 1` نیازی به بررسی جداگانهٔ وجود کلید نیست.

</div>

```python
text = "to be or not to be that is the question"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

- `get` دسترسی امن بدون `KeyError` می‌دهد و `in` وجود کلید را بررسی می‌کند.
