## ۱. عملگرهای مقایسه‌ای

عملگرهای مقایسه‌ای دو مقدار را مقایسه می‌کنند و حاصلشان همیشه `True` یا `False` (نوع `bool`) است.

| عملگر | معنی |
|---|---|
| `==` | مساوی |
| `!=` | نامساوی |
| `>` / `<` | بزرگ‌تر / کوچک‌تر |
| `>=` / `<=` | بزرگ‌تر یا مساوی / کوچک‌تر یا مساوی |

> مراقب باشید: `=` انتساب است و `==` مقایسه.

</div>

```python
x = 7
print(x == 7)
print(x != 7)
print(x > 5)
print(x <= 6)
print("abc" == "ABC")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

پایتون اجازه می‌دهد مقایسه‌ها را مثل ریاضی **زنجیر** کنیم:

</div>

```python
age = 25
print(18 <= age < 30)
```

## ۲. عملگرهای منطقی

برای ترکیب چند شرط از عملگرهای منطقی استفاده می‌کنیم:

| عملگر | نتیجه `True` است وقتی… |
|---|---|
| `and` | **هر دو** شرط درست باشند |
| `or` | **دست‌کم یکی** از شرط‌ها درست باشد |
| `not` | شرط **نادرست** باشد (برعکس می‌کند) |

</div>

```python
age = 20
has_license = True
print(age >= 18 and has_license)
print(age >= 18 and not has_license)
print(age < 18 or has_license)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۳. دستور if

ساختار `if` یک بلوک کد را **فقط وقتی شرط درست است** اجرا می‌کند:

- بعد از شرط **دونقطه `:`** می‌آید.
- خطوط داخل بلوک باید **تورفتگی (indentation)** داشته باشند؛ قرارداد استاندارد **چهار فاصله** است.
- تورفتگی در پایتون اختیاری نیست؛ مرز بلوک‌ها را مشخص می‌کند.

</div>

```python
temperature = 35
if temperature > 30:
    print("It's hot!")
    print("Drink water.")
print("This line always runs")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

اگر تورفتگی را فراموش کنیم، خطای `IndentationError` می‌گیریم:

</div>

```python
if 5 > 2:
print("five is greater")
```

## ۴. else و elif

- `else` بلوکی است که وقتی شرط `if` **نادرست** باشد اجرا می‌شود.
- `elif` (کوتاه‌شدهٔ else if) شرط دیگری را بررسی می‌کند. می‌توانیم هر تعداد `elif` داشته باشیم.

پایتون شرط‌ها را **از بالا به پایین** بررسی می‌کند و **فقط اولین** بلوکی که شرطش درست است اجرا می‌شود.

</div>

```python
score = float(input("Your score (0-20): "))
if score >= 10:
    print("Passed")
else:
    print("Failed")
```

```python
score = 14.5
if score >= 17:
    grade = "A"
elif score >= 14:
    grade = "B"
elif score >= 12:
    grade = "C"
elif score >= 10:
    grade = "D"
else:
    grade = "F"
print(f"Score {score} -> grade {grade}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

مثال: تشخیص زوج یا فرد بودن عدد ورودی:

</div>

```python
n = int(input("Enter an integer: "))
if n % 2 == 0:
    print(f"{n} is even")
else:
    print(f"{n} is odd")
```

## ۵. شرط‌های تودرتو

می‌توانیم یک `if` را داخل `if` دیگر بنویسیم. ولی تودرتو کردن زیاد خوانایی را کم می‌کند؛ اغلب می‌توان با
`and` یا `elif` آن را ساده‌تر کرد.

</div>

```python
age = 22
is_student = True

if age >= 18:
    if is_student:
        print("Adult student: 30% discount")
    else:
        print("Adult: no discount")
else:
    print("Under 18: 50% discount")
```

## ۶. درستی و نادرستی مقدارها (Truthiness)

در شرط می‌توان هر مقداری را گذاشت، نه فقط `bool`. مقدارهای **«نادرست‌گونه»** عبارت‌اند از:
`False`، `0`، `0.0`، رشتهٔ خالی `""`، `None` و ظرف‌های خالی مثل `[]`. بقیهٔ مقدارها «درست‌گونه» هستند.

</div>

```python
name = ""
if name:
    print("Hello", name)
else:
    print("Name is empty")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۷. عبارت شرطی یک‌خطی

برای انتخاب ساده بین دو مقدار، شکل کوتاهی وجود دارد: `value_if_true if condition else value_if_false`

</div>

```python
score = 14
status = "pass" if score >= 10 else "fail"
print(status)
```

## ۸. پروژهٔ کوچک: ماشین‌حساب ساده

برنامه‌ای که دو عدد و یک عملگر بگیرد و نتیجه را حساب کند. به حالت تقسیم بر صفر و عملگر نامعتبر هم توجه شده است.

</div>

```python
x = float(input("First number: "))
op = input("Operator (+ - * /): ")
y = float(input("Second number: "))

if op == "+":
    print(f"{x} + {y} = {x + y}")
elif op == "-":
    print(f"{x} - {y} = {x - y}")
elif op == "*":
    print(f"{x} * {y} = {x * y}")
elif op == "/":
    if y == 0:
        print("Error: division by zero")
    else:
        print(f"{x} / {y} = {x / y}")
else:
    print("Invalid operator")
```
