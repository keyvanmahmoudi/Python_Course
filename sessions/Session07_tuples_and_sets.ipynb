## ۱. تاپل

**تاپل (tuple)** مثل لیست، دنباله‌ای مرتب است ولی **تغییرناپذیر** است: بعد از ساخت نمی‌توان عضوی را عوض، اضافه یا حذف کرد.
تاپل با پرانتز `()` ساخته می‌شود.

**چه زمانی تاپل؟** وقتی داده نباید تغییر کند و ساختار ثابتی دارد؛ مثل مختصات `(x, y)`، رنگ `(R, G, B)`
یا اطلاعات ثابت یک رکورد. تاپل کمی سریع‌تر از لیست است و (برخلاف لیست) می‌تواند کلید دیکشنری باشد.

</div>

```python
point = (3, 4)
rgb = (255, 128, 0)
date = (1405, 8, 17)

print(point[0], point[1])
print(rgb[-1])
print(date[:2])
print(len(rgb))
```

```python
point = (3, 4)
point[0] = 10
```

<div dir="rtl">

تاپل یک‌عضوی حتماً باید ویرگول داشته باشد؛ وگرنه پرانتز فقط پرانتز ریاضی است:

</div>

```python
a = (5,)
b = (5)
print(type(a), type(b))
```

<div dir="rtl">

تاپل دو متد بیشتر ندارد: `count` و `index`. با `tuple()` و `list()` می‌توان بین این دو تبدیل کرد:

</div>

```python
t = (1, 2, 2, 3, 2)
print(t.count(2), t.index(3))
print(list(t))
print(tuple([10, 20]))
```

## ۲. باز کردن تاپل (Unpacking)

می‌توانیم اعضای یک تاپل (یا هر دنباله‌ای) را مستقیم در چند متغیر بریزیم. تعداد متغیرها باید با تعداد اعضا برابر باشد.
با `*` می‌توانیم «بقیهٔ اعضا» را در یک لیست جمع کنیم.

در واقع جابه‌جایی `a, b = b, a` که قبلاً دیدیم هم با همین مکانیزم کار می‌کند.

</div>

```python
student = ("Sara", 21, "Babol")
name, age, city = student
print(name)
print(age)
print(city)
```

```python
first, *rest = [10, 20, 30, 40]
print(first, rest)

*others, last = [10, 20, 30, 40]
print(others, last)
```

<div dir="rtl">

unpacking در حلقه هم بسیار کاربرد دارد:

</div>

```python
products = [("pen", 45000), ("book", 280000), ("bag", 950000)]
for name, price in products:
    print(f"{name:<6}{price:>10,}")
```

## ۳. مجموعه (set)

**مجموعه** گردآیه‌ای **بدون ترتیب** و **بدون عضو تکراری** است و با آکولاد `{}` ساخته می‌شود.

- چون ترتیب ندارد، **اندیس ندارد**.
- اعضای تکراری خودکار حذف می‌شوند؛ پس `set(list)` سریع‌ترین راه حذف تکراری‌هاست.
- بررسی عضویت (`x in s`) در مجموعه بسیار سریع‌تر از لیست است.
- مجموعهٔ خالی را با `set()` می‌سازیم؛ `{}` یک **دیکشنری** خالی است.

</div>

```python
names = ["Ali", "Sara", "Ali", "Reza", "Sara"]
unique_names = set(names)
print(unique_names)
print(len(names), "->", len(unique_names))
```

```python
skills = {"python", "sql"}
skills.add("git")
skills.add("python")      # already there: nothing happens
print(skills)

skills.remove("sql")
skills.discard("java")    # not there: no error
print(skills)
print("git" in skills)
```

```python
empty_dict = {}
empty_set = set()
print(type(empty_dict), type(empty_set))
```

## ۴. عملیات مجموعه‌ها

مجموعه‌ها از عملیات ریاضی مجموعه‌ها پشتیبانی می‌کنند:

| عملیات | عملگر | متد | نتیجه |
|---|---|---|---|
| اجتماع | `a \| b` | `a.union(b)` | همهٔ اعضای هر دو |
| اشتراک | `a & b` | `a.intersection(b)` | اعضای مشترک |
| تفاضل | `a - b` | `a.difference(b)` | در `a` ولی نه در `b` |
| تفاضل متقارن | `a ^ b` | `a.symmetric_difference(b)` | فقط در یکی از دو |

</div>

```python
python_class = {"Ali", "Sara", "Reza", "Mina"}
sql_class = {"Sara", "Mina", "Hamed"}

print("both classes:   ", python_class & sql_class)
print("either class:   ", python_class | sql_class)
print("only python:    ", python_class - sql_class)
print("exactly one:    ", python_class ^ sql_class)
```

```python
required = {"python", "sql"}
my_skills = {"python", "sql", "git", "linux"}
print(required.issubset(my_skills))
print(required <= my_skills)
```

## ۵. کدام ساختار داده؟

| ویژگی | لیست `[]` | تاپل `()` | مجموعه `{}` |
|---|---|---|---|
| ترتیب دارد | ✅ | ✅ | ❌ |
| تغییرپذیر | ✅ | ❌ | ✅ |
| عضو تکراری | ✅ | ✅ | ❌ |
| اندیس | ✅ | ✅ | ❌ |
| کاربرد نمونه | فهرست خرید، نمرات | مختصات، رکورد ثابت | برچسب‌ها، حذف تکراری |

قاعدهٔ سرانگشتی: پیش‌فرض **لیست**؛ اگر داده نباید تغییر کند **تاپل**؛ اگر یکتایی یا عملیات مجموعه مهم است **مجموعه**.

</div>
