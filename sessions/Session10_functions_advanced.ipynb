## ۱. آرگومان‌های متغیر: *args

گاهی نمی‌دانیم تابع چند آرگومان می‌گیرد؛ مثل `print` که هر تعداد مقدار را می‌پذیرد.
پارامتری که با `*` شروع شود (معمولاً `*args`) همهٔ آرگومان‌های موقعیتی اضافه را در یک **تاپل** جمع می‌کند.

</div>

```python
def average(*numbers):
    print("received:", numbers)
    return sum(numbers) / len(numbers)

print(average(10, 20))
print(average(14, 17, 19, 20))
```

<div dir="rtl">

برعکس، با `*` می‌توان یک لیست را هنگام فراخوانی «باز» کرد و اعضایش را به‌عنوان آرگومان‌های جدا فرستاد:

</div>

```python
scores = [15, 18, 12]
print(average(*scores))
```

<div dir="rtl">

## ۲. آرگومان‌های نام‌دار متغیر: **kwargs

پارامتری که با `**` شروع شود (معمولاً `**kwargs`) همهٔ آرگومان‌های **نام‌دار** اضافه را در یک **دیکشنری** جمع می‌کند.
این الگو را در کتابخانه‌هایی مثل matplotlib و scikit-learn زیاد می‌بینید.

</div>

```python
def profile(name, **info):
    print(name)
    for key, value in info.items():
        print(f"  {key}: {value}")

profile("Ali", age=21, city="Amol")
profile("Sara", major="AI", gpa=18.7, year=3)
```

## ۳. حوزهٔ متغیرها (Scope)

متغیری که **داخل** تابع ساخته می‌شود **محلی (local)** است: فقط داخل همان تابع وجود دارد و بعد از پایان تابع از بین می‌رود.
متغیری که **بیرون** همهٔ توابع ساخته شود **سراسری (global)** است و داخل توابع قابل **خواندن** است.

اگر داخل تابع به نامی مقدار بدهیم، پایتون یک متغیر محلی جدید می‌سازد و متغیر سراسری هم‌نام دست نمی‌خورد.
کلمهٔ `global` این رفتار را عوض می‌کند، اما استفاده از آن معمولاً نشانهٔ طراحی ضعیف است؛ بهتر است ورودی را
با پارامتر بگیریم و نتیجه را با `return` برگردانیم.

</div>

```python
x = 10

def f():
    x = 99
    y = 5
    print("inside: x =", x, "y =", y)

f()
print("outside: x =", x)
```

```python
print(y)
```

```python
rate = 0.2

def apply_rate(amount):
    return amount * rate   # reading a global is fine

print(apply_rate(1000))
```

## ۴. توابع lambda

`lambda` یک تابع **کوچک و بی‌نام** در یک خط می‌سازد: `lambda parameters: expression`.
بدنهٔ آن فقط یک عبارت است که خودکار برگردانده می‌شود.

کاربرد اصلی `lambda` وقتی است که باید یک تابع کوتاه را **به تابع دیگری بفرستیم**؛ مثل پارامتر `key` در
`sorted`، `max` و `min`. در پایتون توابع هم مقدارند و می‌توان آن‌ها را مثل هر مقدار دیگری فرستاد.

</div>

```python
square = lambda x: x ** 2
print(square(6))
```

```python
students = [("Ali", 17), ("Sara", 19), ("Reza", 14), ("Mina", 18)]

print(sorted(students))
print(sorted(students, key=lambda s: s[1]))
print(sorted(students, key=lambda s: s[1], reverse=True))
print(max(students, key=lambda s: s[1]))
```

<div dir="rtl">

توابع `map` و `filter` هم یک تابع و یک دنباله می‌گیرند. در پایتون مدرن معمولاً به‌جای آن‌ها از comprehension (بخش بعد) استفاده می‌شود، ولی در کدهای دیگران زیاد می‌بینیدشان:

</div>

```python
prices = [100, 250, 80, 400]
print(list(map(lambda p: p * 1.1, prices)))
print(list(filter(lambda p: p > 90, prices)))
```

## ۵. List Comprehension

**List Comprehension** روشی کوتاه و خوانا برای ساختن لیست از روی یک دنباله است. به جای سه چهار خط
(لیست خالی، حلقه، `append`) یک خط می‌نویسیم:

`[expression for item in iterable if condition]`

بخش `if` اختیاری است و فقط اعضایی را نگه می‌دارد که شرط را دارند.

</div>

<div dir="rtl">

مقایسهٔ روش معمولی و comprehension:

</div>

```python
squares = []
for x in range(1, 6):
    squares.append(x ** 2)
print(squares)

squares = [x ** 2 for x in range(1, 6)]
print(squares)
```

```python
scores = [12, 18, 9, 16, 7, 20]
passed = [s for s in scores if s >= 10]
print(passed)

names = ["  ali ", "SARA", "reza  "]
clean = [n.strip().title() for n in names]
print(clean)
```

<div dir="rtl">

در بخش عبارت می‌توان از عبارت شرطی یک‌خطی هم استفاده کرد (دقت کنید این `if ... else` **قبل** از `for` می‌آید):

</div>

```python
scores = [12, 18, 9, 16]
labels = ["pass" if s >= 10 else "fail" for s in scores]
print(labels)
```

## ۶. Dict و Set Comprehension

همین ایده برای دیکشنری و مجموعه هم وجود دارد:

- دیکشنری: `{key_expr: value_expr for item in iterable}`
- مجموعه: `{expr for item in iterable}`

</div>

```python
words = ["python", "ai", "data", "learning"]
lengths = {w: len(w) for w in words}
print(lengths)

prices = {"pen": 45000, "book": 280000, "bag": 950000}
discounted = {item: p * 0.8 for item, p in prices.items()}
print(discounted)

initials = {name[0] for name in ["Ali", "Amir", "Sara", "Reza", "Sina"]}
print(initials)
```

<div dir="rtl">

> 💡 **نکته:** comprehension را برای منطق ساده به کار ببرید. اگر شرط‌ها و حلقه‌های تودرتو زیاد شد، حلقهٔ معمولی خواناتر است.

</div>

## ۷. آشنایی با توابع بازگشتی

تابع **بازگشتی (recursive)** تابعی است که خودش را صدا می‌زند. هر تابع بازگشتی دو بخش دارد:

1. **حالت پایه:** جایی که بازگشت متوقف می‌شود.
2. **گام بازگشتی:** مسئله را کوچک‌تر می‌کند و تابع را دوباره صدا می‌زند.

مثال کلاسیک، فاکتوریل است: $n! = n \times (n-1)!$ و $0! = 1$.

</div>

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
print(factorial(10))
```
