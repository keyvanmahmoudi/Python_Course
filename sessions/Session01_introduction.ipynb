## ۱. پایتون چیست؟

پایتون (Python) یک زبان برنامه‌نویسی **سطح بالا**، **تفسیری** و **همه‌منظوره** است که در سال ۱۹۹۱ توسط
خیدو فان روسوم (Guido van Rossum) منتشر شد. «سطح بالا» یعنی کد پایتون به زبان انسان نزدیک است و لازم نیست
درگیر جزئیات سخت‌افزار شویم. «تفسیری» یعنی برنامه خط‌به‌خط توسط **مفسر** پایتون اجرا می‌شود و برای دیدن نتیجه
لازم نیست ابتدا کل برنامه را کامپایل کنیم.

**چرا پایتون؟**

- **ساده و خوانا:** دستور زبان پایتون شبیه زبان انگلیسی است و برای شروع برنامه‌نویسی بسیار مناسب است.
- **کتابخانه‌های فراوان:** تقریباً برای هر کاری یک کتابخانهٔ آماده وجود دارد.
- **جامعهٔ بزرگ:** پاسخ تقریباً هر سؤالی را می‌توان در اینترنت پیدا کرد.
- **زبان اصلی هوش مصنوعی و علم داده:** کتابخانه‌هایی مثل NumPy، pandas، scikit-learn و PyTorch همه پایتونی هستند.

**کاربردها:** علم داده و تحلیل داده، یادگیری ماشین و هوش مصنوعی، توسعهٔ وب (Django و Flask)،
خودکارسازی کارهای تکراری، برنامه‌نویسی علمی و ساخت ابزارهای کوچک روزمره.

## ۳. کار با نوت‌بوک Jupyter

نوت‌بوک از **سلول‌ها** ساخته شده است. دو نوع سلول اصلی داریم:

- **سلول کد (Code):** کد پایتون در آن نوشته و اجرا می‌شود و خروجی زیر آن نمایش داده می‌شود.
- **سلول متن (Markdown):** برای نوشتن توضیحات، عنوان‌ها و فرمول‌ها؛ همین متنی که در حال خواندنش هستید.

**میان‌برهای مهم:**

| کلید | کار |
|---|---|
| `Shift + Enter` | اجرای سلول و رفتن به سلول بعدی |
| `Ctrl + Enter` | اجرای سلول و ماندن در همان سلول |
| `Esc` سپس `A` / `B` | افزودن سلول جدید بالا / پایین |
| `Esc` سپس `M` / `Y` | تبدیل سلول به متن / کد |

اگر آخرین خط یک سلول کد یک **مقدار** باشد، Jupyter آن را خودکار نمایش می‌دهد. سلول زیر را اجرا کنید:

## ۴. اولین برنامه: تابع print

تابع `print` هر چه داخل پرانتزش بنویسیم را در خروجی چاپ می‌کند. متن باید داخل **کوتیشن**
(`"..."` یا `'...'`) نوشته شود؛ به متن در برنامه‌نویسی **رشته (string)** می‌گوییم.

</div>

```python
print("Hello, World!")
print("سلام دنیا")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

می‌توانیم چند مقدار را با ویرگول جدا کنیم. `print` بین آن‌ها یک فاصله می‌گذارد. اعداد را بدون کوتیشن می‌نویسیم:

</div>

```python
print("Python", 3, "is fun")
print("2 + 3 =", 2 + 3)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

`print` دو پارامتر کاربردی دارد:

- `sep`: جداکنندهٔ بین مقدارها (پیش‌فرض یک فاصله)
- `end`: چیزی که در انتهای خروجی چاپ می‌شود (پیش‌فرض رفتن به خط بعد)

</div>

```python
print("2026", "11", "08", sep="/")
print("first line", end=" | ")
print("still the same line")
```

## ۷. خواندن پیام‌های خطا

اشتباه کردن بخش طبیعی برنامه‌نویسی است. وقتی کد خطا دارد، پایتون یک **پیام خطا** نشان می‌دهد که
نوع خطا و محل آن را می‌گوید. مهم‌ترین مهارت یک برنامه‌نویس، **خواندن دقیق آخرین خط پیام خطا** است.

سه خطای رایج برای شروع:

| خطا | علت معمول |
|---|---|
| `SyntaxError` | قواعد نوشتاری رعایت نشده؛ مثلاً کوتیشن یا پرانتز بسته نشده |
| `NameError` | از نامی استفاده شده که تعریف نشده یا اشتباه تایپ شده |
| `IndentationError` | تورفتگی (فاصلهٔ اول خط) نادرست است |

سلول زیر عمداً خطا دارد (`Print` با حرف بزرگ نوشته شده). آن را اجرا کنید و پیام خطا را بخوانید:

</div>

```python
Print("Hello")
```

</div>
