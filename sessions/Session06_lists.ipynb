## ۱. ساخت لیست

لیست با کروشه `[]` ساخته می‌شود و اعضا با ویرگول از هم جدا می‌شوند. ویژگی‌های لیست:

- **مرتب (ordered):** ترتیب اعضا حفظ می‌شود.
- **تغییرپذیر (mutable):** می‌توان اعضا را عوض، اضافه یا حذف کرد.
- **ناهمگن:** اعضا می‌توانند از نوع‌های مختلف باشند (هرچند معمولاً هم‌نوع‌اند).
- عضو تکراری مجاز است.

</div>

```python
scores = [18, 15.5, 20, 12, 17]
fruits = ["apple", "banana", "cherry"]
mixed = [1, "two", 3.0, True]
empty = []

print(scores)
print(fruits)
print(mixed)
print(len(scores), len(empty))
```

## ۲. اندیس و برش

اندیس‌گذاری و برش لیست **دقیقاً مثل رشته** است: اندیس از صفر شروع می‌شود، اندیس منفی از انتها می‌شمارد
و `lst[start:stop:step]` بخشی از لیست را به شکل یک **لیست جدید** برمی‌گرداند.

</div>

```python
fruits = ["apple", "banana", "cherry", "date", "fig"]
print(fruits[0])
print(fruits[-1])
print(fruits[1:4])
print(fruits[::-1])
print("cherry" in fruits, "kiwi" in fruits)
print(fruits.index("date"))
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۳. تغییر، افزودن و حذف

برخلاف رشته، لیست **تغییرپذیر** است:

| عملیات | کار |
|---|---|
| `lst[i] = x` | تغییر عضو `i` |
| `append(x)` | افزودن به انتها |
| `insert(i, x)` | افزودن در محل `i` |
| `extend(other)` | افزودن همهٔ اعضای لیست دیگر |
| `remove(x)` | حذف اولین `x` |
| `pop()` / `pop(i)` | حذف و برگرداندن عضو آخر / عضو `i` |
| `del lst[i]` | حذف عضو `i` |
| `clear()` | خالی کردن لیست |

> این متدها لیست را **در جا** تغییر می‌دهند و مقداری برنمی‌گردانند (جز `pop`).

</div>

```python
cart = ["pen", "book"]
cart.append("bag")
print(cart)

cart.insert(0, "phone")
print(cart)

cart.extend(["ruler", "eraser"])
print(cart)

cart[1] = "pencil"
print(cart)
```

```python
cart.remove("book")
print(cart)

last = cart.pop()
print("popped:", last, "->", cart)

del cart[0]
print(cart)
```

> ⚠️ **هشدار:** اشتباه رایج: `cart = cart.append("x")`. چون `append` چیزی برنمی‌گرداند (`None`)، با این کار لیست را از دست می‌دهید.
فقط بنویسید `cart.append("x")`.

## ۴. مرتب‌سازی و توابع آماری

- متد `sort()` خود لیست را مرتب می‌کند؛ تابع `sorted(lst)` یک **لیست جدید مرتب** برمی‌گرداند و اصلی دست نمی‌خورد.
- با `reverse=True` مرتب‌سازی نزولی می‌شود.
- توابع `sum`، `min`، `max` و `len` روی لیست عددی بسیار کاربردی‌اند.

</div>

```python
scores = [14, 19.5, 11, 17, 15.5]
print("sorted copy:", sorted(scores))
print("original:   ", scores)

scores.sort(reverse=True)
print("sorted in place (desc):", scores)
```

```python
scores = [14, 19.5, 11, 17, 15.5]
print("count:", len(scores))
print("sum:", sum(scores))
print("min:", min(scores), "max:", max(scores))
print(f"average: {sum(scores) / len(scores):.2f}")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

لیست رشته‌ها به ترتیب الفبایی مرتب می‌شود. با پارامتر `key` می‌توانیم معیار مرتب‌سازی را عوض کنیم؛ مثلاً بر اساس طول:

</div>

```python
words = ["python", "AI", "data", "learning"]
print(sorted(words))
print(sorted(words, key=len))
```

## ۵. پیمایش لیست با حلقه

حلقهٔ `for` روی اعضای لیست حرکت می‌کند. الگوی رایج: ساختن یک لیست جدید از روی لیست قدیمی با `append`.

</div>

```python
scores = [12, 18, 9, 16, 7, 20]
passed = []
for score in scores:
    if score >= 10:
        passed.append(score)
print("passed:", passed)
```

```python
names = ["Ali", "Sara", "Reza"]
for i, name in enumerate(names, start=1):
    print(i, name)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

تابع `zip` چند لیست را کنار هم قرار می‌دهد تا بتوانیم هم‌زمان رویشان حرکت کنیم:

</div>

```python
names = ["Ali", "Sara", "Reza"]
scores = [17, 19, 14]
for name, score in zip(names, scores):
    print(f"{name:<6}{score}")
```

## ۶. ارجاع در برابر کپی

این یکی از مهم‌ترین نکات پایتون است: دستور `b = a` **کپی نمی‌سازد**؛ فقط یک نام دیگر برای **همان لیست** می‌سازد.
پس تغییر `b` روی `a` هم دیده می‌شود. برای کپی واقعی از `a.copy()` یا `a[:]` استفاده می‌کنیم.

</div>

```python
a = [1, 2, 3]
b = a
b.append(4)
print("a =", a)
print("b =", b)
print(a is b)
```

```python
a = [1, 2, 3]
c = a.copy()
c.append(4)
print("a =", a)
print("c =", c)
print(a is c)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

## ۷. لیست‌های تودرتو

عضو لیست می‌تواند خودش لیست باشد. این ساختار برای نمایش داده‌های **جدولی** (سطر و ستون) مناسب است.
با دو اندیس `matrix[row][col]` به هر خانه دسترسی داریم.

</div>

```python
grades = [
    [18, 17, 19],   # Ali
    [15, 16, 14],   # Sara
    [20, 19, 18],   # Reza
]
print(grades[1])
print(grades[1][2])

for row in grades:
    print(row, "average:", round(sum(row) / len(row), 2))
```
