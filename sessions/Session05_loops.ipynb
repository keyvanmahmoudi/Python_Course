## ۱. حلقهٔ while

حلقهٔ `while` تا وقتی **شرطش درست است** بلوک داخلی را تکرار می‌کند. در هر دور، ابتدا شرط بررسی و
اگر درست بود بلوک اجرا می‌شود.

> مهم: داخل حلقه باید چیزی تغییر کند که بالاخره شرط را نادرست کند؛ وگرنه حلقه **بی‌پایان** می‌شود.
> اگر چنین شد، در Jupyter دکمهٔ توقف (■) را بزنید.

</div>

```python
count = 1
while count <= 5:
    print("count =", count)
    count += 1
print("Loop finished")
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

مثال: چند سال طول می‌کشد تا سپرده‌ای با سود سالانهٔ ۲۵٪ دو برابر شود؟ تعداد تکرار از قبل معلوم نیست؛ این جای `while` است:

</div>

```python
balance = 100_000_000
target = 2 * balance
years = 0
while balance < target:
    balance *= 1.25
    years += 1
print(f"After {years} years: {balance:,.0f} Toman")
```

## ۲. حلقهٔ for و تابع range

حلقهٔ `for` روی اعضای یک **دنباله** (مثل رشته، لیست یا `range`) یکی‌یکی حرکت می‌کند.
تابع `range` دنباله‌ای از اعداد صحیح می‌سازد:

| فراخوانی | اعداد |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(2, 6)` | `2, 3, 4, 5` |
| `range(1, 10, 2)` | `1, 3, 5, 7, 9` |
| `range(10, 0, -3)` | `10, 7, 4, 1` |

مثل برش، **عدد پایانی شامل نمی‌شود**.

</div>

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 11, 3):
    print(i, end=" ")
print()
for i in range(10, 0, -3):
    print(i, end=" ")
```

## ۳. پیمایش رشته و enumerate

`for` می‌تواند مستقیم روی کاراکترهای یک رشته حرکت کند. اگر به **شماره** هر عضو هم نیاز داشتیم،
از `enumerate` استفاده می‌کنیم که در هر دور یک جفت (اندیس، مقدار) برمی‌گرداند.

</div>

```python
for ch in "Python":
    print(ch, end=" ")
```

```python
word = "banana"
vowels = 0
for ch in word:
    if ch in "aeiou":
        vowels += 1
print(f"'{word}' has {vowels} vowels")
```

```python
for i, ch in enumerate("AI", start=1):
    print(i, ch)
```

## ۴. الگوی جمع‌کننده (Accumulator)

یکی از رایج‌ترین الگوها: متغیری را قبل از حلقه مقداردهی اولیه کنیم و در هر دور به‌روزش کنیم.
مثلاً جمع، حاصل‌ضرب، شمارش یا پیدا کردن بیشینه.

</div>

```python
total = 0
for i in range(1, 101):
    total += i
print("Sum 1..100 =", total)

factorial = 1
for i in range(1, 6):
    factorial *= i
print("5! =", factorial)
```

## ۵. break و continue

- `break`: حلقه را **فوراً** تمام می‌کند و از آن بیرون می‌رود.
- `continue`: بقیهٔ دور فعلی را رد می‌کند و **به دور بعد** می‌رود.

الگوی رایج `while True` همراه با `break` برای وقتی است که شرط پایان، وسط حلقه مشخص می‌شود؛ مثلاً
گرفتن ورودی تا وقتی که کاربر مقدار خاصی وارد کند.

</div>

```python
for i in range(1, 10):
    if i % 2 == 0:
        continue
    if i > 7:
        break
    print(i)
```

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

مثال: گرفتن عدد از کاربر تا وقتی که عدد منفی وارد کند، و محاسبهٔ میانگین اعداد واردشده:

</div>

```python
total = 0
count = 0
while True:
    x = float(input("Enter a number (negative to stop): "))
    if x < 0:
        break
    total += x
    count += 1

if count > 0:
    print(f"{count} numbers, average = {total / count:.2f}")
```

## ۶. حلقه‌های تودرتو

حلقه می‌تواند داخل حلقهٔ دیگر باشد. به ازای **هر دور** حلقهٔ بیرونی، حلقهٔ درونی **کامل** اجرا می‌شود.

</div>

```python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()
```

```python
for i in range(1, 5):
    print("*" * i)
```

## ۷. پروژهٔ کوچک: بازی حدس عدد

کامپیوتر عددی تصادفی بین ۱ تا ۱۰۰ انتخاب می‌کند و کاربر حداکثر ۷ بار فرصت حدس دارد. بعد از هر حدس،
راهنمایی «بزرگ‌تر» یا «کوچک‌تر» داده می‌شود. ماژول `random` را در جلسهٔ ۱۱ کامل می‌بینیم؛
`random.seed` باعث می‌شود عدد تصادفی در هر اجرا یکسان باشد تا نتیجه قابل تکرار شود.

در این مثال از ساختار `for ... else` هم استفاده کرده‌ایم: بلوک `else` حلقه فقط وقتی اجرا می‌شود که حلقه
**بدون `break`** تمام شود.

</div>

```python
import random

random.seed(42)
secret = random.randint(1, 100)
max_tries = 7

for attempt in range(1, max_tries + 1):
    guess = int(input(f"Guess #{attempt}: "))
    if guess == secret:
        print(f"Correct! You found it in {attempt} tries.")
        break
    elif guess < secret:
        print("Go higher")
    else:
        print("Go lower")
else:
    print(f"Out of tries. The number was {secret}.")
```

- `for` روی اعضای یک دنباله حرکت می‌کند و `range(start, stop, step)` اعداد را می‌سازد.

- الگوی جمع‌کننده: مقداردهی قبل از حلقه، به‌روزرسانی داخل حلقه.
