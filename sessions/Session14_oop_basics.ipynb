## ۱. کلاس و شیء

- **کلاس (class)** یک **نقشه یا الگو** است؛ مثل نقشهٔ ساختمان.
- **شیء (object)** یا **نمونه (instance)** یک ساختهٔ واقعی از روی آن نقشه است؛ مثل خود ساختمان.
  از یک کلاس می‌توان هر تعداد شیء ساخت.

هر شیء دو چیز دارد:

- **ویژگی‌ها (attributes):** داده‌های شیء؛ مثلاً نام و موجودی یک حساب بانکی.
- **متدها (methods):** کارهایی که شیء انجام می‌دهد؛ مثلاً واریز و برداشت.

در واقع از ابتدای دوره با شیء کار کرده‌ایم: `"hello"` یک شیء از کلاس `str` است و `upper()` یکی از متدهای آن.

</div>

````python
text = "hello"
numbers = [3, 1, 2]
print(type(text), type(numbers))
print(text.upper())
numbers.sort()
print(numbers)
````

## ۲. تعریف کلاس

کلاس با کلمهٔ کلیدی `class` تعریف می‌شود. نام کلاس به سبک **PascalCase** (حرف اول هر کلمه بزرگ) نوشته می‌شود.

متد ویژهٔ `__init__` (**سازنده**) هنگام ساخت هر شیء خودکار اجرا می‌شود و ویژگی‌های اولیه را مقداردهی می‌کند.

پارامتر اول هر متد، `self`، به **خود شیء** اشاره دارد. با `self.name = name` ویژگی `name` را روی شیء ذخیره می‌کنیم.
هنگام فراخوانی، `self` را نمی‌فرستیم؛ پایتون خودش آن را می‌فرستد.

</div>

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score


s1 = Student("Ali", 17)
s2 = Student("Sara", 19)

print(s1.name, s1.score)
print(s2.name, s2.score)
print(type(s1))
````

<div dir="rtl" style="font-family: 'Vazirmatn FD', Vazirmatn, Tahoma, sans-serif;">

هر شیء ویژگی‌های مستقل خودش را دارد و می‌توان آن‌ها را تغییر داد:

</div>

````python
s1.score = 18
print(s1.score, s2.score)
````

## ۳. متدها

متد تابعی است که داخل کلاس تعریف می‌شود و اولین پارامترش `self` است. متد می‌تواند ویژگی‌های شیء را بخواند یا تغییر دهد.

</div>

````python
class BankAccount:
    """A simple bank account."""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit amount must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("insufficient funds")
        self.balance -= amount

    def show(self):
        print(f"{self.owner}: {self.balance:,} Toman")


account = BankAccount("Keyvan", 1_000_000)
account.deposit(500_000)
account.withdraw(200_000)
account.show()
````

````python
try:
    account.withdraw(5_000_000)
except ValueError as e:
    print("Error:", e)
account.show()
````

## ۴. ویژگی کلاس و ویژگی نمونه

- **ویژگی نمونه (instance attribute):** با `self.x` در `__init__` تعریف می‌شود و برای هر شیء جداست.
- **ویژگی کلاس (class attribute):** مستقیم در بدنهٔ کلاس تعریف می‌شود و بین **همهٔ** اشیا مشترک است؛
  مثلاً نرخ سود یا شمارندهٔ تعداد اشیای ساخته‌شده.

</div>

````python
class Product:
    tax_rate = 0.1
    count = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.count += 1

    def price_with_tax(self):
        return self.price * (1 + Product.tax_rate)


p1 = Product("pen", 45_000)
p2 = Product("book", 280_000)

print(Product.count)
print(p1.price_with_tax(), p2.price_with_tax())

Product.tax_rate = 0.09
print(p1.price_with_tax())
````

## ۵. نمایش شیء با __str__

اگر شیئی را `print` کنیم، به‌طور پیش‌فرض چیزی مثل `<__main__.Student object at 0x...>` می‌بینیم که مفید نیست.
متد ویژهٔ `__str__` تعیین می‌کند شیء هنگام `print` و `str()` چطور نمایش داده شود.

متدهایی که با دو زیرخط شروع و تمام می‌شوند **متدهای ویژه (dunder methods)** نام دارند و پایتون در موقعیت‌های خاص
خودش آن‌ها را صدا می‌زند. در جلسهٔ بعد چند مورد دیگر را می‌بینیم.

</div>

````python
class Student:
    def __init__(self, name, scores=None):
        self.name = name
        self.scores = scores if scores is not None else []

    def add_score(self, score):
        if not 0 <= score <= 20:
            raise ValueError("score must be between 0 and 20")
        self.scores.append(score)

    def average(self):
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def __str__(self):
        return f"{self.name} (average: {self.average():.2f})"


ali = Student("Ali")
for sc in [18, 16.5, 19]:
    ali.add_score(sc)
print(ali)
print(ali.scores)
````
