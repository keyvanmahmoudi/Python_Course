# جلسهٔ ۹: توابع (۱)

وقتی یک کار را چند بار در برنامه انجام می‌دهیم، به‌جای تکرار کد آن را در یک **تابع** می‌گذاریم.
تابع کد را منظم، قابل استفادهٔ مجدد و قابل آزمایش می‌کند. در این جلسه تعریف و فراخوانی تابع،

- مقدار پیش‌فرض پارامترها
- برگرداندن مقدار با `return` و تفاوتش با `print`

## ۱. چرا تابع؟

تا اینجا از توابع آمادهٔ زیادی استفاده کرده‌ایم: `print`، `len`، `sum`، `input`. حالا می‌خواهیم توابع خودمان را بسازیم.

## ۲. تعریف و فراخوانی تابع

تابع با کلمهٔ کلیدی `def` تعریف می‌شود. بعد از نام تابع، پرانتز و دونقطه می‌آید و بدنهٔ تابع با تورفتگی نوشته می‌شود.
تعریف تابع به‌تنهایی کاری انجام نمی‌دهد؛ باید آن را **فراخوانی** کنیم.

</div>

```python
def show_banner():
    print("=" * 30)
    print("   Python Course - Session 9")
    print("=" * 30)

show_banner()
```

<div dir="rtl">

## ۳. پارامتر و آرگومان

**پارامتر** متغیری است که در تعریف تابع می‌آید؛ **آرگومان** مقداری است که هنگام فراخوانی می‌فرستیم.

آرگومان‌ها را به دو شکل می‌توان فرستاد:

- **موقعیتی (positional):** بر اساس ترتیب: `describe("Ali", 21)`
- **نام‌دار (keyword):** با ذکر نام پارامتر، ترتیب مهم نیست: `describe(age=21, name="Ali")`

</div>

```python
def describe(name, age):
    print(f"{name} is {age} years old")

describe("Ali", 21)
describe(age=23, name="Sara")
```

</div>

## ۴. مقدار بازگشتی با return

`return` نتیجهٔ تابع را به فراخواننده **برمی‌گرداند** تا بتوانیم آن را در متغیر بگذاریم یا در محاسبهٔ دیگری استفاده کنیم.
با رسیدن به `return`، اجرای تابع فوراً تمام می‌شود.

**تفاوت مهم `print` و `return`:** `print` فقط چیزی را روی صفحه نشان می‌دهد؛ `return` مقدار را برای استفاده
در ادامهٔ برنامه تحویل می‌دهد. تابعی که `return` ندارد، `None` برمی‌گرداند.

</div>

```python
def square(n):
    return n ** 2

result = square(4)
print(result)
print(square(3) + square(4))
```

```python
def square_print(n):
    print(n ** 2)

value = square_print(5)
print("returned value:", value)
```

<div dir="rtl">

تابع می‌تواند چند `return` داشته باشد؛ اولین `return`ی که اجرا شود کار تابع را تمام می‌کند:

</div>

```python
def grade(score):
    if score >= 17:
        return "A"
    if score >= 14:
        return "B"
    if score >= 10:
        return "C"
    return "F"

for sc in [19, 15, 11, 6]:
    print(sc, grade(sc))
```

<div dir="rtl">

برای برگرداندن چند مقدار، آن‌ها را با ویرگول جدا می‌کنیم؛ پایتون یک تاپل برمی‌گرداند که می‌توان آن را unpack کرد:

</div>

```python
def min_max_avg(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, avg = min_max_avg([12, 18, 15, 20, 9])
print(low, high, avg)
```

## ۵. مقدار پیش‌فرض پارامترها

می‌توانیم برای پارامتر مقدار پیش‌فرض تعیین کنیم؛ اگر آرگومانش فرستاده نشود، مقدار پیش‌فرض استفاده می‌شود.
پارامترهای دارای پیش‌فرض باید **بعد از** پارامترهای بدون پیش‌فرض بیایند.

</div>

```python
def final_price(price, discount=0.1, tax=0.1):
    after_discount = price * (1 - discount)
    return after_discount * (1 + tax)

print(final_price(1_000_000))
print(final_price(1_000_000, discount=0.3))
print(final_price(1_000_000, tax=0))
```

<div dir="rtl">

> ⚠️ **هشدار:** هرگز از لیست یا دیکشنری به‌عنوان مقدار پیش‌فرض استفاده نکنید (`def f(items=[])`)؛ آن لیست بین همهٔ فراخوانی‌ها مشترک
می‌شود. به‌جایش از `None` استفاده کنید و داخل تابع لیست را بسازید.

</div>

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("a"))
print(add_item("b"))
```

## ۶. مستندسازی با docstring

رشته‌ای که درست در اولین خط بدنهٔ تابع می‌آید **docstring** نام دارد و توضیح می‌دهد تابع چه می‌کند،
چه ورودی‌هایی می‌گیرد و چه برمی‌گرداند. تابع `help` و ویرایشگرها این توضیح را نمایش می‌دهند.

</div>

```python
def average(numbers):
    """Return the arithmetic mean of a non-empty list of numbers."""
    return sum(numbers) / len(numbers)

help(average)
print(average([14, 16, 18]))
```

## ۷. تابعی که تابع دیگری را صدا می‌زند

توابع می‌توانند از هم استفاده کنند. شکستن یک مسئلهٔ بزرگ به چند تابع کوچک و ساده، مهم‌ترین مهارت طراحی برنامه است.

</div>

```python
def average(numbers):
    """Return the mean of a list of numbers."""
    return sum(numbers) / len(numbers)


def letter_grade(score):
    """Map a 0-20 score to a letter grade."""
    if score >= 17:
        return "A"
    elif score >= 14:
        return "B"
    elif score >= 10:
        return "C"
    return "F"


def report(name, scores):
    """Print a one-line report for a student."""
    avg = average(scores)
    print(f"{name:<6} avg={avg:5.2f} grade={letter_grade(avg)}")


report("Ali", [18, 17, 19])
report("Sara", [12, 15, 13])
report("Reza", [9, 8, 11])
```
